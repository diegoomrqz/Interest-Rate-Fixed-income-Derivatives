# Ois dicounting

**OIS discounting** consiste en descontar los flujos de una operación utilizando factores de descuento obtenidos de una curva calibrada con instrumentos ligados a una tasa *overnight*. Su fundamento está en la remuneración del colateral y en cómo esta afecta el financiamiento de la operación.

Para desarrollarlo, supongamos una operación con colateral en efectivo, en la misma moneda que sus pagos, ajustado continuamente para cubrir todo su valor. Denotemos por $c_t$ la tasa a la que se remunera ese colateral. Este es el caso idealizado de **colateralización perfecta**.

**1. Por qué importa la tasa del colateral**

Supón que tu posición tiene valor positivo $V_t$. Bajo estos supuestos, recibes colateral por ese monto y debes remunerarlo a la tasa $c_t$.

En el modelo estándar, con posibilidad de reutilizar el efectivo recibido, esa remuneración entra en el costo de mantener y replicar la posición. El resultado es que la tasa relevante para descontar sus flujos es $c_t$.

Definimos la cuenta de acumulación asociada,

$$
B_t^c=\exp\left(\int_0^t c_s\,ds\right).
$$

En un modelo donde esta cuenta puede utilizarse como numerario y existe su medida de valuación asociada \(\mathbb Q^c\), el valor de un pago $X_T$ es

$$
\boxed{
V_t
=
B_t^c\,
\mathbb E^{\mathbb Q^c}
\left[
\frac{X_T}{B_T^c}
\middle|\mathcal F_t
\right]
=
\mathbb E^{\mathbb Q^c}
\left[
e^{-\int_t^T c_s\,ds}X_T
\middle|\mathcal F_t
\right].
}
$$

Esto permite definir el factor de descuento colateralizado,

$$
\boxed{
P^c(t,T)
=
\mathbb E^{\mathbb Q^c}
\left[
e^{-\int_t^T c_s\,ds}
\middle|\mathcal F_t
\right].
}
$$

Para un pago conocido \(C\),

$$
V_t=C\,P^c(t,T).
$$

Si \(c_s\) es determinista en \([t,T]\), la expresión se simplifica a

$$
P^c(t,T)=e^{-\int_t^T c_s\,ds}.
$$

Cuando es estocástica, conservamos la esperanza condicional.

**2. Cómo aparece el OIS**

Un *Overnight Indexed Swap* intercambia una pata fija por una pata ligada a la **capitalización de tasas overnight**.

Consideremos un periodo \([T_{i-1},T_i]\), dividido en intervalos diarios con fracciones de año \(\delta_{i,j}\). Si \(r_{i,j}^{\mathrm{ON}}\) es la tasa overnight aplicable a cada intervalo, el cupón flotante pagado al final del periodo es

$$
\boxed{
C_i^{\mathrm{ON}}
=
N\left[
\prod_{j=1}^{m_i}
\left(1+\delta_{i,j}r_{i,j}^{\mathrm{ON}}\right)-1
\right].
}
$$

Su tasa anualizada equivalente es

$$
R_i^{\mathrm{ON}}
=
\frac{
\displaystyle\prod_{j=1}^{m_i}
\left(1+\delta_{i,j}r_{i,j}^{\mathrm{ON}}\right)-1
}{
\alpha_i
},
$$

donde \(\alpha_i\) es la fracción de año del periodo completo.

Así, el cupón puede escribirse como

$$
C_i^{\mathrm{ON}}=N\alpha_iR_i^{\mathrm{ON}}.
$$

La pata fija paga \(NK\alpha_i\) cuando usamos las mismas fracciones de año para simplificar.

**3. La conexión entre la tasa overnight y el descuento**

Supongamos ahora que la tasa de remuneración del colateral coincide con la tasa overnight del OIS. En la idealización continua, el factor de acumulación flotante de un periodo es

$$
\frac{B_{T_i}^c}{B_{T_{i-1}}^c}
=
e^{\int_{T_{i-1}}^{T_i}c_s\,ds}.
$$

Por tanto, el pago flotante es

$$
N\left(
\frac{B_{T_i}^c}{B_{T_{i-1}}^c}-1
\right).
$$

Para \(t\le T_{i-1}\), su valor presente es

$$
\begin{aligned}
V_{t,i}^{\mathrm{flotante}}
&=
N B_t^c\,
\mathbb E^{\mathbb Q^c}
\left[
\frac{1}{B_{T_i}^c}
\left(
\frac{B_{T_i}^c}{B_{T_{i-1}}^c}-1
\right)
\middle|\mathcal F_t
\right]\\
&=
N B_t^c\,
\mathbb E^{\mathbb Q^c}
\left[
\frac{1}{B_{T_{i-1}}^c}
-
\frac{1}{B_{T_i}^c}
\middle|\mathcal F_t
\right].
\end{aligned}
$$

Reconociendo los factores de descuento,

$$
\boxed{
V_{t,i}^{\mathrm{flotante}}
=
N\left[
P^c(t,T_{i-1})-P^c(t,T_i)
\right].
}
$$

La misma cancelación funciona en tiempo discreto cuando la cuenta de colateral acumula exactamente con los factores diarios del cupón.

**4. Valuación de un OIS y construcción de su curva**

Consideremos un OIS que comienza en \(T_0\ge t\), termina en \(T_n\) y paga cada cupón al final de su periodo, sin desfases de pago ni ajustes de observación.

Al sumar los cupones flotantes,

$$
V_t^{\mathrm{flotante}}
=
N\sum_{i=1}^{n}
\left[
P^c(t,T_{i-1})-P^c(t,T_i)
\right].
$$

La suma telescópica da

$$
\boxed{
V_t^{\mathrm{flotante}}
=
N\left[P^c(t,T_0)-P^c(t,T_n)\right].
}
$$

La pata fija vale

$$
V_t^{\mathrm{fija}}
=
NK\sum_{i=1}^{n}\alpha_iP^c(t,T_i).
$$

La tasa par que hace que el swap tenga valor inicial cero es

$$
\boxed{
K_{\mathrm{OIS}}
=
\frac{
P^c(t,T_0)-P^c(t,T_n)
}{
\displaystyle\sum_{i=1}^{n}\alpha_iP^c(t,T_i)
}.
}
$$

Esta es la ecuación que conecta **cotizaciones OIS observadas** con **factores de descuento desconocidos**.

Si \(T_0=t\), entonces \(P^c(t,T_0)=1\). Si además conocemos todos los descuentos anteriores al último pago, podemos despejar

$$
\boxed{
P^c(t,T_n)
=
\frac{
1-K_{\mathrm{OIS}}
\displaystyle\sum_{i=1}^{n-1}\alpha_iP^c(t,T_i)
}{
1+K_{\mathrm{OIS}}\alpha_n
}.
}
$$

Es el mismo mecanismo de bootstrapping que ya desarrollamos, ahora aplicado a instrumentos cuyo índice flotante es overnight. Las convenciones reales de fechas, observación y pago deben incorporarse al calibrar cada instrumento.

**5. Qué significa usar esa curva para descontar otra operación**

Una vez calibrada la curva, podemos valuar pagos deterministas \(C_i\) sujetos al mismo régimen de colateral,

$$
\boxed{
V_t=\sum_{i=1}^{n}C_iP^c(t,T_i).
}
$$

Por ejemplo, si una operación paga \(1{,}000{,}000\) en un año y la curva da

$$
P^c(0,1)=0.96,
$$

su valor presente es

$$
V_0=1{,}000{,}000(0.96)=960{,}000.
$$

Para un cupón flotante futuro, el monto todavía es aleatorio. Usando la medida forward asociada al vencimiento \(T_i\), su valor es

$$
\boxed{
V_{t,i}
=
P^c(t,T_i)\,
\mathbb E^{\mathbb Q^{c,T_i}}
\left[C_i\mid\mathcal F_t\right].
}
$$

Aquí aparecen las dos tareas de una arquitectura multicurva. **La curva de descuento determina \(P^c(t,T_i)\)**. La proyección del índice determina el cupón esperado bajo la medida correspondiente.

La identificación entre descuento colateralizado y descuento OIS se apoya en que la remuneración contractual del colateral coincide con la tasa overnight utilizada para construir la curva. La moneda y las condiciones del acuerdo de colateral determinan qué curva corresponde usar.

# Curvas de Proyección

Siguen las **curvas de proyección** (*projection curves*). Su función es determinar las tasas forward utilizadas para calcular los cupones flotantes futuros de un índice específico.

La curva OIS nos dio los factores de descuento. Ahora necesitamos determinar **qué monto esperamos pagar o recibir en cada fecha**.

**El cupón que queremos valuar**

Consideremos un índice de tasa a plazo $x$. Para un periodo $[T_{i-1},T_i]$, denotemos por

$$
L^x(T_{i-1};T_{i-1},T_i)
$$

la tasa que se fija al comienzo del periodo y se paga al final. Si el nominal es $N$ y la fracción de año es $\delta_i$, el cupón es

$$
C_i=N\delta_iL^x(T_{i-1};T_{i-1},T_i).
$$

Para \(t<T_{i-1}\), ese monto es aleatorio.

Supongamos el mismo régimen de colateralización perfecta que usamos antes, con factores de descuento \($P^c(t,T)$\). Bajo la medida forward asociada al vencimiento de pago $T_i$,

$$
V_{t,i}
=
P^c(t,T_i)
\mathbb E^{\mathbb Q^{c,T_i}}
\left[C_i\mid\mathcal F_t\right].
$$

Sustituyendo el cupón,

$$
V_{t,i}
=
N\delta_iP^c(t,T_i)
\mathbb E^{\mathbb Q^{c,T_i}}
\left[
L^x(T_{i-1};T_{i-1},T_i)
\mid\mathcal F_t
\right].
$$

**Definición de la tasa forward de proyección**

Definimos

$$
\boxed{
F^{x,c}(t;T_{i-1},T_i)
:=
\mathbb E^{\mathbb Q^{c,T_i}}
\left[
L^x(T_{i-1};T_{i-1},T_i)
\mid\mathcal F_t
\right].
}
$$

El superíndice \($x$\) identifica el índice proyectado. El superíndice \($c$\) hace explícito el régimen de colateral y descuento.

Con esta definición,

$$
\boxed{
V_{t,i}
=
N\delta_iP^c(t,T_i)
F^{x,c}(t;T_{i-1},T_i).
}
$$

La fórmula separa las dos funciones de la arquitectura multicurva.

| Componente | Función |
|---|---|
| \($P^c(t,T_i)$\) | Descontar el pago de \($T_i$\) hasta \($t$\) |
| \($F^{x,c}(t;T_{i-1},T_i)$\) | Proyectar el índice del cupón |
| \($N\delta_i$\) | Convertir la tasa anualizada en monto |

Esta esperanza está tomada bajo una **medida de valuación**. Su interpretación como forward se obtiene del precio de contratos, mientras que un pronóstico estadístico requeriría especificar la dinámica bajo la medida física.

**Qué cambia respecto al modelo de una sola curva**

En el modelo de una sola curva, la tasa del índice se identifica con la tasa simple implícita en los bonos cupón cero,

$$
L(T_{i-1};T_{i-1},T_i)
=
\frac{1}{\delta_i}
\left[
\frac{1}{P(T_{i-1},T_i)}-1
\right].
$$

Esa identificación permite obtener

$$
F(t;T_{i-1},T_i)
=
\frac{1}{\delta_i}
\left[
\frac{P(t,T_{i-1})}{P(t,T_i)}-1
\right].
$$

En multicurva, un índice \($x$\) tiene su propia definición contractual y sus propias cotizaciones. Por ello, en general,

$$
\boxed{
F^{x,c}(t;T_{i-1},T_i)
\ne
\frac{1}{\delta_i}
\left[
\frac{P^c(t,T_{i-1})}{P^c(t,T_i)}-1
\right].
}
$$

El lado derecho es el forward simple implícito en la **curva de descuento**. El lado izquierdo corresponde al **índice del contrato**.

**Cómo se representa una curva de proyección**

Podemos almacenar directamente los forwards en sus periodos correspondientes. Otra representación habitual utiliza factores positivos auxiliares $P^{x,c}(t,T)$, llamados **pseudofactores de descuento**, definidos para satisfacer

$$
\boxed{
F^{x,c}(t;T_{i-1},T_i)
=
\frac{1}{\delta_i}
\left[
\frac{P^{x,c}(t,T_{i-1})}
{P^{x,c}(t,T_i)}
-1
\right].
}
$$

Para una sucesión de periodos consecutivos, fijada una normalización inicial, se construyen recursivamente,

$$
P^{x,c}(t,T_i)
=
\frac{P^{x,c}(t,T_{i-1})}
{1+\delta_iF^{x,c}(t;T_{i-1},T_i)}.
$$

Se requiere que los factores de acumulación sean positivos.

Estos pseudofactores son una **parametrización de los forwards**. Para utilizarlos como precios de activos negociables o como numerarios haría falta una justificación adicional. En la valuación del cupón, el descuento sigue siendo \($P^c(t,T_i)$\).

**Cómo se calibra la curva de proyección**

Primero fijamos la curva de descuento $P^c$. Después utilizamos cotizaciones de instrumentos ligados al índice $6yy7u $para determinar los parámetros de proyección.

Consideremos un swap con nominal \$N$, pagos fijos en $S_1,\ldots,S_m$, pagos flotantes en $T_1,\ldots,T_n$ y tasa fija par cotizada $K$.

La pata fija vale

$$
V_t^{\mathrm{fija}}
=
NK\sum_{j=1}^{m}\alpha_jP^c(t,S_j).
$$

La pata flotante vale

$$
V_t^{\mathrm{flotante}}
=
N\sum_{i=1}^{n}
\delta_iP^c(t,T_i)
F^{x,c}(t;T_{i-1},T_i).
$$

La condición de calibración es

$$
\boxed{
K\sum_{j=1}^{m}\alpha_jP^c(t,S_j)
=
\sum_{i=1}^{n}
\delta_iP^c(t,T_i)
F^{x,c}(t;T_{i-1},T_i).
}
$$

En esta ecuación conocemos la tasa par \(K\), los calendarios y los descuentos. Las incógnitas pertenecen a la curva de proyección.

Si los primeros \(n-1\) forwards ya están determinados y el último es la única incógnita, podemos despejarlo,

$$
\boxed{
F_n^{x,c}
=
\frac{
K\displaystyle\sum_{j=1}^{m}\alpha_jP^c(t,S_j)
-
\displaystyle\sum_{i=1}^{n-1}
\delta_iP^c(t,T_i)F_i^{x,c}
}{
\delta_nP^c(t,T_n)
}.
}
$$

Es un bootstrapping de forwards. Cuando una cotización involucra varios periodos aún desconocidos, la parametrización e interpolación deben reducir el problema a suficientes incógnitas identificables.

**6. Ejemplo numérico**

Supongamos un swap de un año con pagos semestrales en ambas patas. Tomemos datos ilustrativos,

$$
P^c(0,0.5)=0.98,
\qquad
P^c(0,1)=0.955,
$$

$$
\delta_1=\delta_2=\alpha_1=\alpha_2=0.5.
$$

La tasa par cotizada es \(K=5\%\), y el primer forward del índice ya se determinó en \(F_1^{x,c}=4.8\%\).

La ecuación par es

$$
0.05\left[0.5(0.98)+0.5(0.955)\right]
=
0.5(0.98)(0.048)
+
0.5(0.955)F_2^{x,c}.
$$

Por tanto,

$$
F_2^{x,c}
=
\frac{0.048375-0.02352}{0.4775}
\approx0.05205236.
$$

Así obtenemos

$$
\boxed{
F_2^{x,c}\approx5.20524\%.
}
$$

El forward simple del mismo periodo calculado exclusivamente con la curva de descuento sería

$$
F_2^c
=
\frac{1}{0.5}
\left(\frac{0.98}{0.955}-1\right)
\approx5.23560\%.
$$

En este ejemplo, las cotizaciones del índice exigen proyectar el segundo cupón al \(5.20524\%\), mientras que su valor presente se calcula usando el descuento \(0.955\).

**7. Qué ocurre cuando llega la fecha de fijación**

Una vez observado el fixing del índice, el cupón queda determinado. Si \(T_{i-1}\le t<T_i\),

$$
\boxed{
V_{t,i}
=
N\delta_i
L^x(T_{i-1};T_{i-1},T_i)
P^c(t,T_i).
}
$$

Para ese cupón sustituimos la proyección por el fixing conocido. Los cupones posteriores siguen utilizando sus forwards.

La derivación anterior supone un índice a plazo fijado al comienzo y pagado al final del periodo. Para un cupón overnight compuesto, la cantidad que se proyecta es el producto de factores diarios. Si el periodo ya comenzó, esa proyección incorpora tanto los fixings observados como las tasas futuras aún desconocidas.

## Basis spreads 

En una arquitectura multicurva, permiten relacionar las curvas de proyección de distintos índices mediante cotizaciones de instrumentos que intercambian una tasa flotante por otra.

Un **basis swap** intercambia dos patas flotantes. El *basis spread* par es el margen que se añade a una de ellas para que el contrato tenga valor inicial cero.

**1. Qué se intercambia**

Consideremos dos índices \(x\) e \(y\), expresados en la misma moneda, bajo un mismo régimen de colateralización.

Supongamos que el contrato establece que recibimos el índice \(x\) más un margen fijo \(s\), y pagamos el índice \(y\).

Los pagos recibidos son

$$
N\alpha_i\left(L_i^x+s\right),
\qquad i=1,\ldots,n,
$$

en fechas \($T_i$\). Los pagos realizados son

$$
N\beta_jL_j^y,
\qquad j=1,\ldots,m,
$$

en fechas \($U_j$\).

Aquí,

- $N$ es el nominal común.
- $\alpha_i$ y \($\beta_j$\) son las fracciones de año respectivas.
- $L_i^x$ y \($L_j^y$\) son los fixings de los índices.
- $s$ es un margen anualizado, constante durante el contrato.

Las patas pueden tener calendarios y frecuencias diferentes. Por ejemplo, un índice a tres meses frente a otro a seis meses.

Supondremos que el margen se añade de forma simple al cupón, de modo que su contribución es \($N\alpha_i s$\).

**Valuación con curvas de proyección separadas**

Sea \($P^c(t,T)$\) el factor de descuento determinado por el régimen de colateral. Denotemos por \($F_i^{x,c}(t)$\) y \($F_j^{y,c}(t)$\) los forwards de proyección correspondientes.

El valor de recibir \($x+s$\) y pagar \($y$\) es

$$
\boxed{
V_t(s)
=
N\sum_{i=1}^{n}
\alpha_iP^c(t,T_i)\left(F_i^{x,c}(t)+s\right)
-
N\sum_{j=1}^{m}
\beta_jP^c(t,U_j)F_j^{y,c}(t).
}
$$

Ambas patas utilizan la misma curva de descuento porque hemos supuesto la misma moneda y el mismo régimen de colateral. Cada índice utiliza su propia curva de proyección.

Para simplificar, definimos los valores de las patas sin margen, por unidad de nominal,

$$
\Phi_x(t)
=
\sum_{i=1}^{n}
\alpha_iP^c(t,T_i)F_i^{x,c}(t),
$$

$$
\Phi_y(t)
=
\sum_{j=1}^{m}
\beta_jP^c(t,U_j)F_j^{y,c}(t).
$$

También definimos la anualidad de la pata que recibe el margen,

$$
A_x(t)
=
\sum_{i=1}^{n}\alpha_iP^c(t,T_i).
$$

Así,

$$
\boxed{
V_t(s)=N\left[\Phi_x(t)+sA_x(t)-\Phi_y(t)\right].
}
$$

**3. Definición formal del basis spread par**

El margen par \($s_t^*$\) satisface

$$
V_t(s_t^*)=0.
$$

Por tanto,

$$
\Phi_x(t)+s_t^*A_x(t)=\Phi_y(t),
$$

y obtenemos

$$
\boxed{
s_t^*
=
\frac{\Phi_y(t)-\Phi_x(t)}{A_x(t)}.
}
$$

Como los factores de descuento y las fracciones de año son positivos,

$$
A_x(t)>0,
$$

por lo que el margen par existe y es único para estas patas y forwards dados.

Su interpretación depende de **qué pata lleva el margen**.

- Si \($\Phi_y>\Phi_x$\), necesitamos \($s_t^*>0$\) para aumentar el valor de la pata \(x\).
- Si \($\Phi_y<\Phi_x$\), el margen sobre \(x\) es negativo.
- Si \($\Phi_y=\Phi_x$\), el margen par es cero.

Al leer una cotización de basis, hay que identificar los dos índices, sus calendarios y la pata a la que se añade el spread.

**4. Relación con las diferencias entre forwards**

Si ambas patas comparten fechas y fracciones de año,

$$
T_i=U_i,
\qquad
\alpha_i=\beta_i,
$$

entonces

$$
s_t^*
=
\frac{
\displaystyle\sum_{i=1}^{n}
\alpha_iP^c(t,T_i)
\left(F_i^{y,c}(t)-F_i^{x,c}(t)\right)
}{
\displaystyle\sum_{i=1}^{n}\alpha_iP^c(t,T_i)
}.
$$

Definiendo

$$
w_i=
\frac{\alpha_iP^c(t,T_i)}{A_x(t)},
$$

tenemos \(w_i>0\) y \(\sum_iw_i=1\). Por ello,

$$
\boxed{
s_t^*
=
\sum_{i=1}^{n}
w_i\left(F_i^{y,c}(t)-F_i^{x,c}(t)\right).
}
$$

En este caso, el basis par es un **promedio ponderado de las diferencias entre forwards**.

Si los calendarios difieren, utilizamos la fórmula general basada en los valores presentes de las patas.

**Ejemplo numérico**

Consideremos dos pagos semestrales con

$$
P^c(0,0.5)=0.98,
\qquad
P^c(0,1)=0.955,
\qquad
\alpha_1=\alpha_2=0.5.
$$

Supongamos los siguientes forwards ilustrativos,

| Periodo | Índice \(x\) | Índice \(y\) |
|---|---:|---:|
| Primer semestre | \(4.80\%\) | \(5.00\%\) |
| Segundo semestre | \(5.20\%\) | \(5.50\%\) |

Los valores de las patas por unidad de nominal son

$$
\Phi_x
=
0.5(0.98)(0.048)+0.5(0.955)(0.052)
=
0.04835,
$$

$$
\Phi_y
=
0.5(0.98)(0.05)+0.5(0.955)(0.055)
=
0.0507625.
$$

La anualidad es

$$
A_x=0.5(0.98)+0.5(0.955)=0.9675.
$$

Por tanto,

$$
s^*
=
\frac{0.0507625-0.04835}{0.9675}
\approx0.00249354.
$$

Un punto base equivale a \(10^{-4}\) en tasa decimal. Así,

$$
\boxed{
s^*\approx24.9354\text{ puntos base}.
}
$$

El intercambio de \(x+24.9354\) puntos base contra \(y\) tiene valor inicial cero bajo estos datos.

Las diferencias entre forwards son 20 y 30 puntos base. El margen par incorpora ambas, ponderadas por sus fracciones de año y descuentos.

**6. Cómo se utiliza el basis para construir otra curva**

Supongamos que ya conocemos

- La curva de descuento \(P^c\).
- La curva de proyección del índice \(x\).
- Una cotización de mercado \(s^{\mathrm{mkt}}\) para intercambiar \(x+s^{\mathrm{mkt}}\) contra \(y\).

La condición de calibración es

$$
\boxed{
\sum_{j=1}^{m}
\beta_jP^c(t,U_j)F_j^{y,c}(t)
=
\Phi_x(t)+s^{\mathrm{mkt}}A_x(t).
}
$$

El lado derecho ya está determinado. Esta ecuación aporta información sobre la curva de proyección de \(y\).

Si conocemos sus primeros \(m-1\) forwards y el último es la única incógnita,

$$
\boxed{
F_m^{y,c}(t)
=
\frac{
\Phi_x(t)+s^{\mathrm{mkt}}A_x(t)
-
\displaystyle\sum_{j=1}^{m-1}
\beta_jP^c(t,U_j)F_j^{y,c}(t)
}{
\beta_mP^c(t,U_m)
}.
}
$$

Así podemos extender la curva \(y\) mediante bootstrapping. Si cada instrumento introduce varios forwards desconocidos, necesitamos una parametrización o interpolación que permita identificarlos.

Una cotización de basis relaciona los valores de ambas patas. Si las dos curvas de proyección fueran completamente desconocidas, harían falta más instrumentos para determinarlas.

**7. Margen contractual y margen par actual**

Al contratar el swap, el margen queda fijado en \(s_0\). Posteriormente, las curvas cambian y el margen que haría par al contrato pasa a ser \(s_t^*\).

Para las mismas patas pendientes y con sus fixings conocidos incorporados cuando corresponda,

$$
V_t(s_0)
=
N\left[\Phi_x(t)+s_0A_x(t)-\Phi_y(t)\right].
$$

Como

$$
\Phi_y(t)-\Phi_x(t)=s_t^*A_x(t),
$$

resulta

$$
\boxed{
V_t(s_0)=NA_x(t)(s_0-s_t^*).
}
$$

Para quien recibe \(x+s_0\) y paga \(y\), un margen contractual superior al margen par actual produce valor positivo.

**8. Qué significa económicamente**

El basis refleja la diferencia de valor entre las exposiciones a los dos índices. Según los índices e instrumentos utilizados, puede recoger diferencias de plazo, crédito, liquidez y condiciones de oferta y demanda.

La colateralización del swap determina su descuento y reduce la exposición frente a la contraparte bajo los supuestos del modelo. Los índices que fijan sus pagos conservan sus propias características económicas.

Por eso podemos tener simultáneamente **una curva común de descuento, distintas curvas de proyección y un basis de mercado que las relaciona**.
