<font size=15 ><b> Riesgo de tasa


</b></font>

El riesgo de tasas de interés es el riesgo de que los cambios en las tasas de mercado modifiquen el valor de un instrumento financiero o de un portafolio.

Esta exposición surge porque las tasas intervienen tanto en el descuento de los pagos futuros como en la determinación de los cupones flotantes. Por ello, un cambio en las tasas puede modificar el valor presente de los pagos y también los montos que esperamos recibir o entregar.

Fijemos la fecha de valuación \(t\) y representemos los factores de tasa mediante un vector

$$\mathbf r=(r_1,\ldots,r_m)^{\mathsf T}$$

El valor de la posición es una función de esos factores,

$$V_t=\mathcal V(t,\mathbf r)$$

Manteniendo constantes los términos del contrato y los demás factores de mercado, un movimiento \($\Delta\mathbf r$\) produce el cambio de valor

$$\boxed{
\Delta V_t
=
\mathcal V(t,\mathbf r+\Delta\mathbf r)
-
\mathcal V(t,\mathbf r)
}$$
Las tasas de distintos vencimientos pueden moverse en cantidades diferentes. En un marco multicurva, también pueden cambiar de manera distinta las curvas de descuento y de proyección.

Para estudiar esa exposición utilizamos las siguientes medidas y conceptos.


| Medida o concepto | Qué describe |
|---|---|
| **DV01/PV01** | Sensibilidad monetaria ante un movimiento de un punto base, según una perturbación especificada |
| **Key-rate DV01** | Sensibilidad monetaria localizada en determinados vencimientos |
| **Duración modificada** | Sensibilidad relativa del precio respecto del rendimiento |
| **Convexidad** | Curvatura de la relación entre precio y tasa |
| **Basis risk** | Exposición a cambios en los diferenciales entre índices o curvas |

# DV01 / PV01

**DV01 / PV01**

El **DV01** mide la sensibilidad monetaria del valor de un instrumento o portafolio ante un movimiento de **un punto base** en la tasa de interés.

Un punto base equivale a

$$
1\text{ pb}=0.01\%=\frac{0.01}{100}=10^{-4}.
$$

Sea \(V(r)\) el valor del portafolio como función de una tasa \(r\), expresada en decimal. Mantenemos constantes la fecha de valuación, los términos contractuales y los demás factores de mercado.

**Derivación a partir del cambio de valor**

Si la tasa pasa de \(r\) a \(r+\Delta r\), el cambio exacto del valor es

$$
\Delta V=V(r+\Delta r)-V(r).
$$

Suponiendo que \(V\) es dos veces continuamente diferenciable alrededor de \(r\), su expansión de Taylor es

$$
V(r+\Delta r)
=
V(r)
+
\frac{\partial V}{\partial r}\Delta r
+
\frac12\frac{\partial^2V}{\partial r^2}(\Delta r)^2
+
o((\Delta r)^2).
$$

Restando \(V(r)\) de ambos lados,

$$
V(r+\Delta r)-V(r)
=
\frac{\partial V}{\partial r}\Delta r
+
\frac12\frac{\partial^2V}{\partial r^2}(\Delta r)^2
+
o((\Delta r)^2).
$$

Por la definición de \(\Delta V\),

$$
\Delta V
=
\frac{\partial V}{\partial r}\Delta r
+
\frac12\frac{\partial^2V}{\partial r^2}(\Delta r)^2
+
o((\Delta r)^2).
$$

Al conservar únicamente el término de primer orden, obtenemos

$$
\boxed{
\Delta V\approx\frac{\partial V}{\partial r}\Delta r.
}
$$

Para una subida de un punto base, \(\Delta r=10^{-4}\). Por tanto,

$$
\Delta V\approx\frac{\partial V}{\partial r}10^{-4}.
$$

Usaremos la convención de DV01 positivo cuando una subida de tasas disminuye el valor,

$$
\boxed{
\operatorname{DV01}
=
-\frac{\partial V}{\partial r}10^{-4}.
}
$$

Para expresar un movimiento cualquiera en puntos base, definimos

$$
\Delta r_{\mathrm{pb}}=\frac{\Delta r}{10^{-4}},
$$

de donde

$$
\Delta r=10^{-4}\Delta r_{\mathrm{pb}}.
$$

Sustituyendo en la aproximación del cambio de valor,

$$
\Delta V
\approx
\frac{\partial V}{\partial r}
10^{-4}\Delta r_{\mathrm{pb}}.
$$

Como

$$
\frac{\partial V}{\partial r}10^{-4}
=
-\operatorname{DV01},
$$

resulta

$$
\boxed{
\Delta V
\approx
-\operatorname{DV01}\,\Delta r_{\mathrm{pb}}.
}
$$

El DV01 se reporta en **unidades monetarias por punto base**.

**Cálculo para flujos fijos**

Consideremos pagos conocidos \(C_i\) en fechas \(T_i\), con plazos en años

$$
\tau_i=T_i-t.
$$

Supongamos una tasa constante \(r\) con capitalización continua. El valor presente es

$$
V(r)=\sum_{i=1}^{n}C_i e^{-r\tau_i}.
$$

Derivamos respecto de \(r\). Como la suma es finita,

$$
\frac{\partial V}{\partial r}
=
\sum_{i=1}^{n}
\frac{\partial}{\partial r}
\left(C_i e^{-r\tau_i}\right).
$$

Los pagos \(C_i\) y los plazos \(\tau_i\) permanecen constantes, por lo que

$$
\frac{\partial V}{\partial r}
=
\sum_{i=1}^{n}
C_i\frac{\partial}{\partial r}e^{-r\tau_i}.
$$

Aplicamos la regla de la cadena,

$$
\frac{\partial}{\partial r}e^{-r\tau_i}
=
e^{-r\tau_i}
\frac{\partial}{\partial r}(-r\tau_i).
$$

Dado que

$$
\frac{\partial}{\partial r}(-r\tau_i)=-\tau_i,
$$

obtenemos

$$
\frac{\partial}{\partial r}e^{-r\tau_i}
=
-\tau_i e^{-r\tau_i}.
$$

Sustituyendo,

$$
\frac{\partial V}{\partial r}
=
\sum_{i=1}^{n}C_i\left(-\tau_i e^{-r\tau_i}\right)
=
-\sum_{i=1}^{n}\tau_iC_i e^{-r\tau_i}.
$$

Finalmente, por la definición del DV01,

$$
\operatorname{DV01}
=
-\left(
-\sum_{i=1}^{n}\tau_iC_i e^{-r\tau_i}
\right)10^{-4},
$$

y por tanto,

$$
\boxed{
\operatorname{DV01}
=
10^{-4}\sum_{i=1}^{n}\tau_iC_i e^{-r\tau_i}.
}
$$

Como \(C_i e^{-r\tau_i}\) es el valor presente del pago \(i\), cada flujo contribuye mediante su **plazo multiplicado por su valor presente**. Para pagos positivos y plazos positivos, la derivada del precio es negativa y el DV01 es positivo.

**Ejemplo**

Un bono cupón cero paga \(N=1{,}000{,}000\) MXN dentro de \(\tau=2\) años. Su valor como función de la tasa continua es

$$
V(r)=Ne^{-r\tau}
=
1{,}000{,}000e^{-2r}.
$$

Derivando paso a paso,

$$
\frac{\partial V}{\partial r}
=
1{,}000{,}000
\frac{\partial}{\partial r}e^{-2r},
$$

$$
\frac{\partial V}{\partial r}
=
1{,}000{,}000e^{-2r}
\frac{\partial}{\partial r}(-2r),
$$

$$
\frac{\partial V}{\partial r}
=
-2(1{,}000{,}000)e^{-2r}.
$$

Evaluando en \(r=0.05\),

$$
\left.\frac{\partial V}{\partial r}\right|_{r=0.05}
=
-2(1{,}000{,}000)e^{-0.10}
\approx-1{,}809{,}674.84.
$$

Por tanto,

$$
\operatorname{DV01}
=
-\left(-1{,}809{,}674.84\right)10^{-4},
$$

$$
\boxed{
\operatorname{DV01}
\approx180.9675\ \mathrm{MXN/pb}.
}
$$

Si la tasa sube de \(5.00\%\) a \(5.10\%\), el cambio en decimal es

$$
\Delta r=0.051-0.05=0.001.
$$

Expresado en puntos base,

$$
\Delta r_{\mathrm{pb}}
=
\frac{0.001}{0.0001}=10.
$$

Entonces,

$$
\Delta V
\approx
-(180.9675)(10),
$$

$$
\boxed{
\Delta V\approx-1{,}809.67\ \mathrm{MXN}.
}
$$

**Relación con PV01**

PV01 también se utiliza para esta sensibilidad del valor presente. Si se define como el cambio firmado ante una subida de un punto base,

$$
\operatorname{PV01}_{+}
=
V(r+10^{-4})-V(r),
$$

la aproximación de primer orden da

$$
\operatorname{PV01}_{+}
\approx
\frac{\partial V}{\partial r}10^{-4}.
$$

Por nuestra definición de DV01,

$$
\boxed{
\operatorname{PV01}_{+}\approx-\operatorname{DV01}.
}
$$

En el ejemplo, el PV01 firmado ante una subida es aproximadamente \(-180.97\) MXN, mientras que el DV01 bajo la convención elegida es \(+180.97\) MXN/pb.

# Key Rate DV01

**Key-rate DV01**

El **key-rate DV01** mide la sensibilidad monetaria del valor de un instrumento o portafolio ante un movimiento de **un punto base en una tasa de un vencimiento específico**, manteniendo constantes las tasas de los demás nodos.

Esto permite identificar **en qué vencimientos se concentra el riesgo**.

Representemos la curva mediante tasas cero en \(m\) vencimientos,

$$
\mathbf r=(r_1,\ldots,r_m)^{\mathsf T},
$$

donde \(r_i\) corresponde al vencimiento \(T_i\). El valor del portafolio es

$$
V=V(r_1,\ldots,r_m).
$$

Fijamos una regla de interpolación para determinar las tasas entre nodos. Así, cambiar \(r_i\) también modifica el tramo de curva que depende de ese nodo.

**Derivación a partir del cambio de valor**

Si las tasas cambian en

$$
\Delta\mathbf r=(\Delta r_1,\ldots,\Delta r_m)^{\mathsf T},
$$

el cambio exacto del valor es

$$
\Delta V
=
V(\mathbf r+\Delta\mathbf r)-V(\mathbf r).
$$

Suponiendo suficiente regularidad, Taylor multivariable da

$$
V(\mathbf r+\Delta\mathbf r)
=
V(\mathbf r)
+
\sum_{i=1}^{m}
\frac{\partial V}{\partial r_i}\Delta r_i
+
O(\|\Delta\mathbf r\|^2).
$$

Restando el valor inicial,

$$
\Delta V
=
\sum_{i=1}^{m}
\frac{\partial V}{\partial r_i}\Delta r_i
+
O(\|\Delta\mathbf r\|^2).
$$

A primer orden,

$$
\boxed{
\Delta V
\approx
\sum_{i=1}^{m}
\frac{\partial V}{\partial r_i}\Delta r_i.
}
$$

Si solamente cambia la tasa del nodo \(k\), entonces

$$
\Delta r_i=0
\qquad\text{para }i\ne k.
$$

Todos esos términos desaparecen y queda

$$
\Delta V
\approx
\frac{\partial V}{\partial r_k}\Delta r_k.
$$

Para una subida de un punto base en ese nodo,

$$
\Delta r_k=10^{-4},
$$

por lo que

$$
\Delta V
\approx
\frac{\partial V}{\partial r_k}10^{-4}.
$$

Manteniendo la convención de signo que usamos para DV01, definimos

$$
\boxed{
\operatorname{KRDV01}_k
=
-\frac{\partial V}{\partial r_k}10^{-4}.
}
$$

Un \(\operatorname{KRDV01}_k\) positivo indica que una subida de la tasa del nodo \(k\) reduce el valor a primer orden.

**Cambio de valor cuando se mueven varios nodos**

Para expresar cada movimiento en puntos base,

$$
\Delta r_{i,\mathrm{pb}}
=
\frac{\Delta r_i}{10^{-4}},
$$

de donde

$$
\Delta r_i
=
10^{-4}\Delta r_{i,\mathrm{pb}}.
$$

Sustituyendo en la expansión,

$$
\Delta V
\approx
\sum_{i=1}^{m}
\frac{\partial V}{\partial r_i}
10^{-4}\Delta r_{i,\mathrm{pb}}.
$$

Como

$$
\frac{\partial V}{\partial r_i}10^{-4}
=
-\operatorname{KRDV01}_i,
$$

obtenemos

$$
\boxed{
\Delta V
\approx
-\sum_{i=1}^{m}
\operatorname{KRDV01}_i\,
\Delta r_{i,\mathrm{pb}}.
}
$$

Cada nodo aporta una ganancia o pérdida aproximada. El cambio total se obtiene sumando esas contribuciones.

**Cálculo de la derivada para flujos fijos**

Consideremos primero el caso en que cada pago \(C_i\) ocurre exactamente en un nodo \(T_i\). Sean

$$
\tau_i=T_i-t
$$

los plazos en años y \(r_i\) las tasas cero continuamente capitalizadas. El valor es

$$
V(r_1,\ldots,r_m)
=
\sum_{i=1}^{m}C_i e^{-r_i\tau_i}.
$$

Para derivar respecto de \(r_k\), separamos su término,

$$
V
=
C_k e^{-r_k\tau_k}
+
\sum_{i\ne k}C_i e^{-r_i\tau_i}.
$$

Los términos con \(i\ne k\) permanecen constantes respecto de \(r_k\). Por tanto,

$$
\frac{\partial V}{\partial r_k}
=
C_k\frac{\partial}{\partial r_k}e^{-r_k\tau_k}.
$$

Aplicando la regla de la cadena,

$$
\frac{\partial}{\partial r_k}e^{-r_k\tau_k}
=
e^{-r_k\tau_k}
\frac{\partial}{\partial r_k}(-r_k\tau_k)
=
-\tau_k e^{-r_k\tau_k}.
$$

Entonces,

$$
\frac{\partial V}{\partial r_k}
=
-\tau_kC_k e^{-r_k\tau_k}.
$$

Sustituyendo en la definición,

$$
\operatorname{KRDV01}_k
=
-\left(-\tau_kC_k e^{-r_k\tau_k}\right)10^{-4},
$$

$$
\boxed{
\operatorname{KRDV01}_k
=
10^{-4}\tau_kC_k e^{-r_k\tau_k}.
}
$$

Si un pago cae entre nodos, su tasa interpolada depende de varios de ellos y su sensibilidad se distribuye entre esos nodos según la interpolación elegida.

**Ejemplo e interpretación**

Supongamos dos pagos positivos, ubicados en los nodos de dos y cinco años, con los siguientes valores presentes,

| Nodo | Valor presente del pago |
|---|---:|
| \(2\) años | \(1{,}000{,}000\) MXN |
| \(5\) años | \(2{,}000{,}000\) MXN |

Sus sensibilidades son

$$
\operatorname{KRDV01}_{2a}
=
10^{-4}(2)(1{,}000{,}000)
=
200\ \mathrm{MXN/pb},
$$

$$
\operatorname{KRDV01}_{5a}
=
10^{-4}(5)(2{,}000{,}000)
=
1{,}000\ \mathrm{MXN/pb}.
$$

Si solamente sube un punto base la tasa a dos años,

$$
\Delta V\approx-200\ \mathrm{MXN}.
$$

Si solamente sube un punto base la tasa a cinco años,

$$
\Delta V\approx-1{,}000\ \mathrm{MXN}.
$$

Por tanto, esta posición es más sensible al nodo de cinco años.

Ahora supongamos que la tasa a dos años sube \(10\) puntos base y la tasa a cinco años baja \(5\) puntos base,

$$
\Delta r_{2a,\mathrm{pb}}=10,
\qquad
\Delta r_{5a,\mathrm{pb}}=-5.
$$

El cambio aproximado es

$$
\Delta V
\approx
-\left[
200(10)+1{,}000(-5)
\right],
$$

$$
\Delta V
\approx
-\left[2{,}000-5{,}000\right],
$$

$$
\boxed{
\Delta V\approx+3{,}000\ \mathrm{MXN}.
}
$$

La pérdida de \(2{,}000\) MXN por el nodo corto queda superada por la ganancia de \(5{,}000\) MXN por el nodo largo.

Cuando todos los nodos suben un punto base y su interpolación reproduce un desplazamiento paralelo, la sensibilidad agregada satisface

$$
\boxed{
\operatorname{DV01}_{\mathrm{paralelo}}
=
\sum_{i=1}^{m}\operatorname{KRDV01}_i.
}
$$

En este ejemplo, el DV01 paralelo es \(200+1{,}000=1{,}200\ \mathrm{MXN/pb}\). Los key-rate DV01 muestran cómo se reparte esa exposición entre los dos vencimientos.

# Duracion modificada 
# $D_{mod}$

**Duración modificada**

La **duración modificada** mide la sensibilidad relativa del precio de un bono ante un cambio pequeño en su rendimiento al vencimiento.

Mientras el DV01 expresa el efecto en unidades monetarias, la duración modificada permite aproximar el cambio del precio **como proporción de su valor inicial**.

**Derivación a partir del cambio de precio**

Sea \(P(y)>0\) el precio completo de un bono como función de su rendimiento \(y\), expresado en decimal.

Ante un cambio pequeño \(\Delta y\), Taylor da

$$
P(y+\Delta y)
=
P(y)
+
\frac{\partial P}{\partial y}\Delta y
+
O((\Delta y)^2).
$$

Restando \(P(y)\),

$$
\Delta P
=
\frac{\partial P}{\partial y}\Delta y
+
O((\Delta y)^2).
$$

A primer orden,

$$
\Delta P\approx\frac{\partial P}{\partial y}\Delta y.
$$

Para obtener el cambio relativo, dividimos entre el precio inicial,

$$
\frac{\Delta P}{P}
\approx
\frac{1}{P}\frac{\partial P}{\partial y}\Delta y.
$$

Definimos la duración modificada como

$$
\boxed{
D_{\mathrm{mod}}
=
-\frac{1}{P}\frac{\partial P}{\partial y}.
}
$$

Por tanto,

$$
\frac{1}{P}\frac{\partial P}{\partial y}
=
-D_{\mathrm{mod}},
$$

y sustituyendo,

$$
\boxed{
\frac{\Delta P}{P}
\approx
-D_{\mathrm{mod}}\Delta y.
}
$$

Para un bono de flujos fijos positivos, la derivada del precio es negativa y la duración modificada es positiva.

**Relación con la duración de Macaulay**

Consideremos flujos fijos \(C_i\), pagaderos dentro de \(\tau_i\) años. Si \(y\) es una tasa nominal anual con \(m\) capitalizaciones por año,

$$
P(y)=\sum_{i=1}^{n}C_i(1+y/m)^{-m\tau_i}.
$$

Derivando respecto de \(y\),

$$
\frac{\partial P}{\partial y}
=
\sum_{i=1}^{n}
C_i(-m\tau_i)(1+y/m)^{-m\tau_i-1}\frac{1}{m}.
$$

Cancelamos \(m\),

$$
\frac{\partial P}{\partial y}
=
-\sum_{i=1}^{n}
\tau_iC_i(1+y/m)^{-m\tau_i-1}.
$$

Extraemos el factor común \((1+y/m)^{-1}\),

$$
\frac{\partial P}{\partial y}
=
-\frac{1}{1+y/m}
\sum_{i=1}^{n}
\tau_iC_i(1+y/m)^{-m\tau_i}.
$$

Sustituyendo esta derivada en la definición,

$$
D_{\mathrm{mod}}
=
-\frac{1}{P}
\left[
-\frac{1}{1+y/m}
\sum_{i=1}^{n}
\tau_iC_i(1+y/m)^{-m\tau_i}
\right].
$$

Los signos negativos se cancelan,

$$
D_{\mathrm{mod}}
=
\frac{1}{1+y/m}
\left[
\frac{
\sum_{i=1}^{n}\tau_iC_i(1+y/m)^{-m\tau_i}
}{P}
\right].
$$

La expresión entre corchetes es la duración de Macaulay. Por tanto,

$$
\boxed{
D_{\mathrm{mod}}
=
\frac{D_{\mathrm{Mac}}}{1+y/m}.
}
$$

Esta relación utiliza el mismo rendimiento y la misma convención de capitalización con los que calculamos el precio y la duración de Macaulay.

**Interpretación y ejemplo**

Retomemos el bono anterior, con nominal \(100\), cupón anual \(10\), vencimiento en dos años y rendimiento anual \(y=10\%\). Obtuvimos

$$
P=100,
\qquad
D_{\mathrm{Mac}}\approx1.909091\text{ años}.
$$

Como \(m=1\),

$$
D_{\mathrm{mod}}
=
\frac{1.909091}{1+0.10}
\approx1.735537.
$$

Supongamos que el rendimiento sube del \(10.00\%\) al \(10.10\%\). El cambio en decimal es

$$
\Delta y=0.101-0.10=0.001.
$$

El cambio relativo aproximado del precio es

$$
\frac{\Delta P}{P}
\approx
-(1.735537)(0.001)
=
-0.001735537.
$$

Para expresarlo como porcentaje, multiplicamos por \(100\),

$$
\boxed{
\frac{\Delta P}{P}\times100
\approx-0.173554\%.
}
$$

El bono pierde aproximadamente el \(0.173554\%\) de su precio inicial. Como \(P=100\),

$$
\Delta P
\approx
100(-0.001735537)
=
-0.173554,
$$

y el nuevo precio aproximado es

$$
\boxed{
P_{\mathrm{nuevo}}
\approx100-0.173554
=
99.826446.
}
$$

Una duración modificada mayor implica un cambio porcentual de mayor magnitud ante el mismo movimiento pequeño del rendimiento.

**Relación con el DV01**

Partimos de

$$
D_{\mathrm{mod}}
=
-\frac{1}{P}\frac{\partial P}{\partial y}.
$$

Multiplicando por \(P\),

$$
PD_{\mathrm{mod}}
=
-\frac{\partial P}{\partial y}.
$$

Multiplicando ahora por \(10^{-4}\),

$$
PD_{\mathrm{mod}}10^{-4}
=
-\frac{\partial P}{\partial y}10^{-4}.
$$

El lado derecho es el DV01 respecto de ese rendimiento. Por tanto,

$$
\boxed{
\operatorname{DV01}
=
P D_{\mathrm{mod}}10^{-4}.
}
$$

En el ejemplo,

$$
\operatorname{DV01}
=
100(1.735537)(10^{-4})
\approx0.0173554.
$$

Una subida de un punto base reduce el precio aproximadamente \(0.0173554\) unidades monetarias por cada \(100\) de nominal.

**Capitalización continua**

Si el precio se expresa mediante un rendimiento continuo \(r\),

$$
P(r)=\sum_{i=1}^{n}C_i e^{-r\tau_i},
$$

entonces

$$
\frac{\partial P}{\partial r}
=
-\sum_{i=1}^{n}\tau_iC_i e^{-r\tau_i}.
$$

Al normalizar por el precio,

$$
D_{\mathrm{mod}}
=
-\frac{1}{P}\frac{\partial P}{\partial r}
=
\frac{\sum_{i=1}^{n}\tau_iC_i e^{-r\tau_i}}{P}.
$$

Esta es la duración de Macaulay calculada con ese rendimiento continuo,

$$
\boxed{
D_{\mathrm{mod}}=D_{\mathrm{Mac}}
\qquad\text{bajo capitalización continua}.
}
$$

# Convexidad

**Convexidad**

La **convexidad** mide la curvatura de la relación entre el precio de un bono y su rendimiento. Permite incorporar el efecto de segundo orden al aproximar el cambio de precio.

La duración modificada utiliza la primera derivada. La convexidad utiliza la **segunda derivada**, normalizada por el precio del bono.

**Derivación a partir del cambio de precio**

Sea \(P(y)>0\) el precio completo de un bono como función de su rendimiento \(y\), expresado en decimal.

Suponiendo que \(P\) es dos veces continuamente diferenciable, para un cambio pequeño \(\Delta y\),

$$
P(y+\Delta y)
=
P(y)
+
\frac{\partial P}{\partial y}\Delta y
+
\frac12\frac{\partial^2P}{\partial y^2}(\Delta y)^2
+
o((\Delta y)^2).
$$

Restando \(P(y)\) y conservando los términos hasta segundo orden,

$$
\Delta P
\approx
\frac{\partial P}{\partial y}\Delta y
+
\frac12\frac{\partial^2P}{\partial y^2}(\Delta y)^2.
$$

Para obtener el cambio relativo, dividimos entre el precio inicial,

$$
\frac{\Delta P}{P}
\approx
\frac{1}{P}\frac{\partial P}{\partial y}\Delta y
+
\frac{1}{2P}\frac{\partial^2P}{\partial y^2}(\Delta y)^2.
$$

Ya definimos la duración modificada como

$$
D_{\mathrm{mod}}
=
-\frac{1}{P}\frac{\partial P}{\partial y}.
$$

Ahora definimos la **convexidad**,

$$
\boxed{
\mathcal C
=
\frac{1}{P}\frac{\partial^2P}{\partial y^2}.
}
$$

Sustituyendo ambas definiciones,

$$
\boxed{
\frac{\Delta P}{P}
\approx
-D_{\mathrm{mod}}\Delta y
+
\frac12\mathcal C(\Delta y)^2.
}
$$

Multiplicando por \(P\), obtenemos la aproximación monetaria,

$$
\boxed{
\Delta P
\approx
-PD_{\mathrm{mod}}\Delta y
+
\frac12P\mathcal C(\Delta y)^2.
}
$$

**Cálculo de la segunda derivada**

Consideremos flujos fijos positivos \(C_i\), pagaderos dentro de \(\tau_i\) años. Si \(y\) es una tasa nominal anual con \(m\) capitalizaciones por año,

$$
P(y)
=
\sum_{i=1}^{n}C_i(1+y/m)^{-m\tau_i},
\qquad 1+\frac ym>0.
$$

Derivando una vez mediante la regla de la cadena,

$$
\frac{\partial P}{\partial y}
=
\sum_{i=1}^{n}
C_i(-m\tau_i)(1+y/m)^{-m\tau_i-1}\frac1m.
$$

Cancelando \(m\),

$$
\frac{\partial P}{\partial y}
=
-\sum_{i=1}^{n}
\tau_iC_i(1+y/m)^{-m\tau_i-1}.
$$

Derivamos nuevamente. Como \(\tau_iC_i\) es constante respecto de \(y\),

$$
\frac{\partial^2P}{\partial y^2}
=
-\sum_{i=1}^{n}
\tau_iC_i
\frac{\partial}{\partial y}
(1+y/m)^{-m\tau_i-1}.
$$

Aplicando la regla de la cadena al término interior,

$$
\frac{\partial}{\partial y}
(1+y/m)^{-m\tau_i-1}
=
(-m\tau_i-1)
(1+y/m)^{-m\tau_i-2}
\frac1m.
$$

Sustituyendo,

$$
\frac{\partial^2P}{\partial y^2}
=
-\sum_{i=1}^{n}
\tau_iC_i
(-m\tau_i-1)
(1+y/m)^{-m\tau_i-2}
\frac1m.
$$

Los signos negativos se cancelan y

$$
\frac{m\tau_i+1}{m}
=
\tau_i+\frac1m.
$$

Por tanto,

$$
\boxed{
\frac{\partial^2P}{\partial y^2}
=
\sum_{i=1}^{n}
\tau_i\left(\tau_i+\frac1m\right)
C_i(1+y/m)^{-m\tau_i-2}.
}
$$

Dividiendo entre \(P\),

$$
\boxed{
\mathcal C
=
\frac1P
\sum_{i=1}^{n}
\frac{
\tau_i(\tau_i+1/m)C_i
}{
(1+y/m)^{m\tau_i+2}
}.
}
$$

Para estos flujos positivos, cada término es positivo. Entonces,

$$
\frac{\partial^2P}{\partial y^2}>0,
\qquad
\mathcal C>0.
$$

**Interpretación**

La primera derivada del precio es negativa, mientras que la segunda es positiva,

$$
\frac{\partial P}{\partial y}<0,
\qquad
\frac{\partial^2P}{\partial y^2}>0.
$$

Esto significa que el precio disminuye cuando aumenta el rendimiento, pero su pendiente se vuelve **menos negativa** conforme el rendimiento sube.

El término de convexidad en la aproximación es

$$
\frac12\mathcal C(\Delta y)^2.
$$

Como \((\Delta y)^2\) es positivo tanto para subidas como para bajadas, este ajuste es positivo en ambos casos.

- Si el rendimiento **sube**, el término de duración produce una pérdida y la convexidad reduce su magnitud.
- Si el rendimiento **baja**, el término de duración produce una ganancia y la convexidad la incrementa.

Por eso, para movimientos pequeños de igual tamaño en sentidos opuestos, un bono de convexidad positiva presenta una ganancia mayor que la pérdida correspondiente.

**Ejemplo**

Retomemos el bono con nominal \(100\), cupón anual de \(10\), vencimiento en dos años y rendimiento anual del \(10\%\),

$$
P(y)=\frac{10}{1+y}+\frac{110}{(1+y)^2}.
$$

En \(y=0.10\),

$$
P=100,
\qquad
D_{\mathrm{mod}}\approx1.735537.
$$

La primera derivada es

$$
P'(y)
=
-10(1+y)^{-2}
-
220(1+y)^{-3}.
$$

Derivando de nuevo,

$$
P''(y)
=
(-10)(-2)(1+y)^{-3}
+
(-220)(-3)(1+y)^{-4},
$$

$$
P''(y)
=
20(1+y)^{-3}
+
660(1+y)^{-4}.
$$

Evaluando en \(y=0.10\),

$$
P''(0.10)
=
\frac{20}{1.10^3}
+
\frac{660}{1.10^4}
\approx465.815177.
$$

Por tanto,

$$
\boxed{
\mathcal C
=
\frac{465.815177}{100}
\approx4.658152.
}
$$

Supongamos una subida del rendimiento del \(10\%\) al \(11\%\),

$$
\Delta y=0.11-0.10=0.01.
$$

El efecto de duración es

$$
-PD_{\mathrm{mod}}\Delta y
=
-100(1.735537)(0.01)
\approx-1.735537.
$$

El ajuste de convexidad es

$$
\frac12P\mathcal C(\Delta y)^2
=
\frac12(100)(4.658152)(0.01)^2.
$$

Como \((0.01)^2=0.0001\),

$$
\frac12P\mathcal C(\Delta y)^2
\approx0.023291.
$$

Sumando ambos efectos,

$$
\Delta P
\approx-1.735537+0.023291
=-1.712246.
$$

El nuevo precio aproximado es

$$
\boxed{
P_{\mathrm{nuevo}}
\approx100-1.712246
=98.287754.
}
$$

La revaluación exacta da

$$
P(0.11)
=
\frac{10}{1.11}
+
\frac{110}{1.11^2}
\approx98.287477.
$$

La duración por sí sola estimaba un precio de \(98.264463\). Incorporar la convexidad aproxima mejor el efecto del movimiento en este ejemplo.

**Capitalización continua**

Si utilizamos un rendimiento continuo \(r\),

$$
P(r)=\sum_{i=1}^{n}C_i e^{-r\tau_i}.
$$

La primera derivada es

$$
\frac{\partial P}{\partial r}
=
-\sum_{i=1}^{n}\tau_iC_i e^{-r\tau_i}.
$$

Derivando de nuevo,

$$
\frac{\partial^2P}{\partial r^2}
=
-\sum_{i=1}^{n}
\tau_iC_i\left(-\tau_i e^{-r\tau_i}\right)
=
\sum_{i=1}^{n}\tau_i^2C_i e^{-r\tau_i}.
$$

Entonces,

$$
\boxed{
\mathcal C
=
\frac{
\sum_{i=1}^{n}\tau_i^2C_i e^{-r\tau_i}
}{
P(r)
}.
}
$$

Con los pesos \(w_i=C_i e^{-r\tau_i}/P(r)\), esta expresión queda

$$
\mathcal C=\sum_{i=1}^{n}w_i\tau_i^2.
$$

Bajo capitalización continua, la convexidad es el promedio ponderado de los **plazos al cuadrado** y se expresa en años cuadrados. Para un bono cupón cero con plazo \(\tau\), todo el peso corresponde a su único pago,

$$
\boxed{\mathcal C=\tau^2.}
$$

# Codigo
Datos y función de precio

El bono tiene nominal de \($1{,}000{,}000$\) MXN, cupones anuales de \($50{,}000$\) MXN y vencimiento en tres años.

Su precio es

$P=\sum_i C_i e^{-r_i\tau_i}.
$
La función acepta una tasa común o una tasa por cada fecha de pago.

```python
import numpy as np

plazos = np.array([1.0, 2.0, 3.0])
flujos = np.array([50_000.0, 50_000.0, 1_050_000.0])

tasa = 0.05
pb = 0.0001


def precio(tasas):
    descuentos = np.exp(-np.asarray(tasas) * plazos)
    return np.sum(flujos * descuentos)


precio_inicial = precio(tasa)

print(f"Precio inicial: {precio_inicial:,.2f} MXN")
```

DV01 / PV01

Conservamos la convención

$$DV01=-P'(r)10^{-4}.
$$
Para nuestros flujos,
$$P'(r)=-\sum_i\tau_iC_i e^{-r\tau_i}.
$$
También calculamos el $DV01$ por diferencia central y el cambio exacto ante una subida de un punto base, al que llamaremos pv01_subida.

```python
def dv01(tasa):
    valores_presentes = flujos * np.exp(-tasa * plazos)
    derivada = -np.sum(plazos * valores_presentes)
    return -derivada * pb


dv01_analitico = dv01(tasa)

# Aproximación de la derivada mediante revaluación.
dv01_numerico = (
    precio(tasa - pb) - precio(tasa + pb)
) / 2

# Cambio exacto ante una subida de 1 pb.
pv01_subida = precio(tasa + pb) - precio_inicial

# Escenario de subida de 10 pb.
cambio_pb = 10
cambio_estimado = -dv01_analitico * cambio_pb
cambio_exacto = precio(tasa + cambio_pb * pb) - precio_inicial

print(f"DV01 analítico: {dv01_analitico:,.4f} MXN/pb")
print(f"DV01 numérico:  {dv01_numerico:,.4f} MXN/pb")
print(f"PV01 firmado ante subida: {pv01_subida:,.4f} MXN")
print(f"Cambio estimado con +10 pb: {cambio_estimado:,.2f} MXN")
print(f"Cambio exacto con +10 pb:   {cambio_exacto:,.2f} MXN")
```

El DV01 es aproximadamente 284.93 MXN/pb. Una subida de diez puntos base produce una pérdida estimada de 2,849.28 MXN, frente a una pérdida exacta de 2,845.10 MXN.

- Key-rate DV01

Cada pago coincide con un nodo de la curva. Por ello,

$$KRDV01_i=10^{-4}\tau_iC_i e^{-r_i\tau_i}.$$

Calculamos las sensibilidades y aplicamos un escenario donde el nodo de un año sube diez puntos base, el de dos años permanece constante y el de tres años baja cinco.

```python
tasas_nodos = np.array([0.05, 0.05, 0.05])


def key_rate_dv01(tasas):
    valores_presentes = flujos * np.exp(-tasas * plazos)
    return pb * plazos * valores_presentes


sensibilidades = key_rate_dv01(tasas_nodos)

for plazo, sensibilidad in zip(plazos, sensibilidades):
    print(f"Nodo {plazo:.0f} años: {sensibilidad:,.4f} MXN/pb")

print(f"Suma de key-rate DV01: {sensibilidades.sum():,.4f} MXN/pb")

movimientos_pb = np.array([10.0, 0.0, -5.0])

contribuciones = -sensibilidades * movimientos_pb
cambio_estimado = contribuciones.sum()

tasas_nuevas = tasas_nodos + movimientos_pb * pb
cambio_exacto = precio(tasas_nuevas) - precio(tasas_nodos)

print(f"Cambio estimado del escenario: {cambio_estimado:,.2f} MXN")
print(f"Cambio exacto del escenario:   {cambio_exacto:,.2f} MXN")
```

Las sensibilidades , son aprox
| Nodo | Key-rate DV01 |
|---|---:|
| Un año | 4.7561 MXN/pb |
| Dos años | 9.0484 MXN/pb |
| Tres años | 271.1230 MXN/pb |

El riesgo se concentra en el pago final. El escenario genera una ganancia estimada de 1,308.05 MXN, frente a 1,309.09 MXN al revaluar.

Duración modificada

Con capitalización continua,

$$D_{\mathrm{mod}}
=
-\frac{P'(r)}{P(r)}
=
\frac{\sum_i\tau_iC_i e^{-r\tau_i}}{P(r)}$$

La utilizamos para estimar el cambio relativo,

$$\frac{\Delta P}{P}\approx-D_{\mathrm{mod}}\Delta r
$$

```python
def duracion_modificada(tasa):
    valores_presentes = flujos * np.exp(-tasa * plazos)
    derivada = -np.sum(plazos * valores_presentes)
    return -derivada / precio(tasa)


duracion = duracion_modificada(tasa)
cambio_tasa = 10 * pb

cambio_relativo_estimado = -duracion * cambio_tasa

cambio_relativo_exacto = (
    precio(tasa + cambio_tasa) - precio_inicial
) / precio_inicial

print(f"Duración modificada: {duracion:.6f}")
print(f"Cambio relativo estimado: {100 * cambio_relativo_estimado:.6f}%")
print(f"Cambio relativo exacto:   {100 * cambio_relativo_exacto:.6f}%")

# Relación con el DV01.
dv01_desde_duracion = precio_inicial * duracion * pb
print(f"DV01 desde duración: {dv01_desde_duracion:,.4f} MXN/pb")
```

La duración modificada es aproximadamente $2.859149$. Una subida de diez puntos base reduce el precio aproximadamente $0.285915 %$.

- Convexidad

Con capitalización continua,

$$\mathcal C
=
\frac{P''(r)}{P(r)}
=
\frac{\sum_i\tau_i^2C_i e^{-r\tau_i}}{P(r)}$$

La aproximación de segundo orden es
$$\Delta P
\approx
-PD_{\mathrm{mod}}\Delta r
+
\frac12P\mathcal C(\Delta r)^2$$

```python
def convexidad(tasa):
    valores_presentes = flujos * np.exp(-tasa * plazos)
    segunda_derivada = np.sum(plazos**2 * valores_presentes)
    return segunda_derivada / precio(tasa)


convexidad_bono = convexidad(tasa)
cambio_tasa = 10 * pb

efecto_duracion = -precio_inicial * duracion * cambio_tasa

efecto_convexidad = (
    0.5 * precio_inicial * convexidad_bono * cambio_tasa**2
)

cambio_segundo_orden = efecto_duracion + efecto_convexidad
cambio_exacto = precio(tasa + cambio_tasa) - precio_inicial

print(f"Convexidad: {convexidad_bono:.6f} años²")
print(f"Efecto de duración:  {efecto_duracion:,.6f} MXN")
print(f"Ajuste convexidad:   {efecto_convexidad:,.6f} MXN")
print(f"Cambio estimado:     {cambio_segundo_orden:,.6f} MXN")
print(f"Cambio exacto:       {cambio_exacto:,.6f} MXN")
```

- Basis risk

Construimos una posición larga en el bono \(A\) y una posición de cobertura en un bono \(B\), ambos con los mismos flujos del ejemplo, pero ligados a dos curvas diferentes.

Para simplificar, representamos cada curva mediante una tasa plana,

$$r_A=5\%,
\qquad
r_B=4.5\%$$
El valor del portafolio es

$$V=P_A(r_A)+q_BP_B(r_B)$$
Elegimos \(q_B\) para neutralizar el DV01 ante movimientos iguales de ambas tasas,

$DV01_A+q_BDV01_B=0$ 
$q_B=-\frac{DV01_A}{DV01_B}$

Después mantenemos fija esa cobertura y aplicamos movimientos diferentes.

```python
tasa_A = 0.05
tasa_B = 0.045

dv01_A = dv01(tasa_A)
dv01_B = dv01(tasa_B)

# Posición negativa significa venta del bono B.
cantidad_B = -dv01_A / dv01_B

dv01_portafolio = dv01_A + cantidad_B * dv01_B

print(f"Posición en A: +1")
print(f"Posición en B: {cantidad_B:.8f}")
print(f"DV01 conjunto: {dv01_portafolio:.8f} MXN/pb")

# Escenario: A sube 12 pb y B sube 8 pb.
movimiento_A_pb = 12
movimiento_B_pb = 8

base_inicial = tasa_A - tasa_B

tasa_A_nueva = tasa_A + movimiento_A_pb * pb
tasa_B_nueva = tasa_B + movimiento_B_pb * pb

base_nueva = tasa_A_nueva - tasa_B_nueva
cambio_base_pb = (base_nueva - base_inicial) / pb

# Aproximación mediante las sensibilidades de ambas posiciones.
cambio_estimado = (
    -dv01_A * movimiento_A_pb
    -cantidad_B * dv01_B * movimiento_B_pb
)

# Como neutralizamos el movimiento común, también podemos usar la base.
cambio_estimado_base = -dv01_A * cambio_base_pb

# Revaluación con la cantidad de cobertura original.
cambio_exacto = (
    precio(tasa_A_nueva) - precio(tasa_A)
    + cantidad_B * (precio(tasa_B_nueva) - precio(tasa_B))
)

print(f"Base inicial: {base_inicial / pb:.2f} pb")
print(f"Base nueva:   {base_nueva / pb:.2f} pb")
print(f"Cambio de base: {cambio_base_pb:.2f} pb")
print(f"Cambio estimado por sensibilidades: {cambio_estimado:,.2f} MXN")
print(f"Cambio estimado por base:           {cambio_estimado_base:,.2f} MXN")
print(f"Cambio exacto:                     {cambio_exacto:,.2f} MXN")
```
