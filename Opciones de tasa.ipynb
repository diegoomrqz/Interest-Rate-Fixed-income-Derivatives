# Caplet 
- Opción sobre la tasa de un periodo
Un caplet es una opción que paga al comprador si la tasa de referencia de un periodo supera una tasa pactada $K$, llamada strike. Sirve, por ejemplo, para proteger un préstamo a tasa variable contra una subida de tasas durante ese periodo.

Consideremos tres fechas $t< T_1< T_2 $:
- En $t$ contratamos el caplet y, en el caso que estudiaremos, pagamos la prima.
- En $T_1$ se observa la tasa de referencia $L_{T_1}$ aplicable al periodo $[T_1,T_2]$. En ese momento queda determinado el importe.
- En $T_2$ se liquida el pago. La fracción de año del periodo es $\delta$ y el nominal sobre el que se calcula el interés es $N$.

Flujo del caplet
Si $L_{T_1}>K$, el interés calculado a la tasa de referencia excede al interés calculado al strike. El caplet paga esa diferencia. Si $L_{T_1}\leq K$, el pago es cero. Por tanto, el flujo del comprador es

$$\boxed{
X_{T_2}^{\mathrm{caplet}}
=
N\delta(L_{T_1}-K)^+

}
$$
Es una call sobre la tasa. El nominal $N$ sólo sirve como base para calcular el pago. El vendedor cobra la prima al contratar y debe pagar el flujo del caplet si la tasa termina por encima de $K$.

## Valuación de un caplet con Black–76

Sea $t<T_1 < T_2$ . En $T_1$se observa la tasa $L_{T_1}$ del periodo $[T_1,T_2]$ y en T_2 se paga
$$
X_{T_2}=N\delta(L_{T_1}-K)^+.
$$
Trabajaremos con una sola curva. Denotamos por $P(u,T)$ el precio en u de un bono cupón cero que paga una unidad en $T$.

Del portafolio del FRA al forward
El portafolio que usamos para el FRA vale

$V_u^{\mathrm{FRA}}(K) = N \left [P(u, T_1 ) - (1+\delta K) P(u,T_2)\right]$

Al igualar ese valor a cero y despejar $K$, obtenemos la tasa forward vigente en $u$:

$F_u=\frac{1}{\delta}
\left[\frac{P(u,T_1)}{P(u,T_2)}-1\right]$.

En el fixing, $$P(T_1,T_1)=1$$, de modo que

$F_{T_1}=\frac{1}{\delta}
\left[\frac{1}{P(T_1,T_2)}-1\right]=L_{T_1}$.


Así que modelar cómo se mueve \(F_u\) hasta \(T_1\) es modelar qué fixing puede terminar entrando al máximo. Para el FRA basta el forward de hoy, porque el flujo es lineal. Para el caplet importa también cómo se distribuyen los posibles fixings a ambos lados del strike.

Así, la variable que entra al payoff es el valor que alcanza el forward en $T_1$. El strike $K$ permanece fijo durante toda la vida del contrato.

El bono de pago como numerario

Elegimos $P(u,T_2)$ como numerario porque el caplet paga en $T_2$. Sea $M_u$ la cuenta monetaria y $\mathbb Q$ su medida neutral al riesgo. El cambio a la medida del bono de $T_2$ viene dado por
$$
Z_u
=\left.\frac{d\mathbb Q^{T_2}}{d\mathbb Q}\right|_{\mathcal F_u}
=
\frac{P(u,T_2)/M_u}{P(0,T_2)/M_0}.
$$

Bajo \mathbb $Q^{T_2}$, el precio del bono de $T_1$ expresado en unidades del bono de $T_2$,

$$R_u=\frac{P(u,T_1)}{P(u,T_2)},
$$
es una martingala. Como $F_u=(R_u-1)/\delta$, el forward también es una martingala bajo esa medida.

En $T_1$ ya sabemos cuánto pagará el caplet. Podemos financiar ese pago comprando $N\delta(F_{T_1}-K)^+$ bonos de $T_2$. Por tanto, en  $T_1$ el caplet vale esa cantidad de unidades del numerario. El valor hoy, en las mismas unidades, es su esperanza condicional:

$$\boxed{
\frac{V_t}{P(t,T_2)}
=
N\delta\,\mathbb E^{\mathbb Q^{T_2}}
\left[(F_{T_1}-K)^+\mid\mathcal F_t\right].
}$$


Dinámica del forward y réplica del caplet
El cambio de numerario determina bajo qué medida el forward es una martingala. Para calcular la esperanza, Black–76 añade el supuesto de volatilidad relativa constante:

$$dF_u=\sigma F_u\,dW_u^{T_2}.
$$

Aquí se ve qué sustituye a las cantidades fijas de bonos del FRA. Escribe $v(u,F_u)=V_u/P(u,T_2)$ para el precio del caplet en unidades del numerario. Una cartera con a_u bonos de $T_1$ y $b_u$ bonos de $T_2$ vale $a_uR_u+b_u$ en esas unidades. Como $dR_u=\delta\,dF_u$ , para igualar el término aleatorio de Itô, $v_F\,dF_u$, elegimos

$$a_u=\frac{v_F(u,F_u)}{\delta},
\qquad
b_u=v(u,F_u)-a_uR_u$$.

Para que la cartera sea autofinanciada, su cambio de valor debe ser $a_u\,dR_u$. Al comparar con Itô, el término en du debe anularse:

$$v_u+\frac12\sigma^2F^2v_{FF}=0,
\qquad
v(T_1,F)=N\delta(F-K)^+$$.

Esta es la réplica en continuo con los mismos dos bonos. En el FRA, el flujo lineal permite mantener cantidades fijas. En el caplet, la cantidad $a_u$ se ajusta a medida que se mueve el forward.

## Fórmula de la prima del caplet
Partimos del precio que ya construimos usando el bono de $T_2$  como numerario:

$$V_t
=
N\delta P(t,T_2)\,
\mathbb E^{\mathbb Q^{T_2}}
\left[(F_{T_1}-K)^+\mid\mathcal F_t\right]$$.

Para obtener la fórmula de Black–76, hay que calcular esa esperanza. Primero encontramos la distribución de $F_{T_1}$ y después integramos el payoff en los estados donde el caplet paga.

1. De la dinámica del forward a la distribución del fixing
Bajo la medida forward \($\mathbb Q^{T_2}$\), Black–76 supone que

$dF_u=\sigma F_u\,dW_u^{T_2},
\qquad t\le u\le T_1$,


donde $W^{T_2}$ es un movimiento browniano bajo esa medida y $\sigma$ es constante. Para aplicar Itô a $g(F_u)=\log F_u$, usamos $g'(F)=1/F$, ($g''(F)=-1/F^2$\) y $d \langle F\rangle_u=\sigma^2F_u^2\,du$.
Entonces

$\begin{aligned}
d\log F_u
&=\frac{1}{F_u}\,dF_u
-\frac{1}{2F_u^2}\,d\langle F\rangle_u\\
&=\sigma\,dW_u^{T_2}-\frac12\sigma^2\,du.
\end{aligned}$

Integramos de $t$ a $T_1$:

$$\log F_{T_1}-\log F_t
=
-\frac12\sigma^2(T_1-t)
+\sigma\left(W_{T_1}^{T_2}-W_t^{T_2}\right)$$.

Sea $\tau=T_1-t$ el tiempo hasta el fixing y $s=\sigma\sqrt{\tau}$. El incremento browniano tiene distribución $\mathcal N(0,\tau)$ y es independiente de $\mathcal F_t$. Por tanto,

$$Z=\frac{W_{T_1}^{T_2}-W_t^{T_2}}{\sqrt{\tau}}
\sim\mathcal N(0,1)$$

bajo $\mathbb Q^{T_2}$. Exponenciando,

$$\boxed{F_{T_1}=F_t e^{-s^2/2+sZ}}$$

Condicionalmente a $\mathcal F_t$, conocemos $F_t$. Toda la incertidumbre que queda en el fixing está representada por $Z$.


2. Encontrar los estados en los que el caplet paga
Para $F_t>0$, $K>0$ y $s>0$, el caplet paga cuando $F_{T_1}>K$. Sustituimos la expresión anterior y despejamos $Z$:

$$\begin{aligned}
F_t e^{-s^2/2+sZ}>K
&\iff -\frac{s^2}{2}+sZ>\log(K/F_t)\\
&\iff Z>\frac{\log(K/F_t)+s^2/2}{s}.
\end{aligned}$$

Definimos

$$ d_2 = \frac{\log(F_t/K)-s^2/2}{s}$$


### 2. Probabilidad de ejercicio del caplet

El caplet paga exactamente cuando $Z>-d_2$. Por simetría de la normal estándar, la probabilidad condicional de ese evento bajo $\mathbb Q^{T_2}$ es

$$
\mathbb Q^{T_2}(F_{T_1}>K\mid\mathcal F_t)
=1-\Phi(-d_2)=\Phi(d_2).
$$

### 3. Integrar el pago en esos estados

Como el payoff es cero para $Z\le -d_2$, sólo integramos de $-d_2$ a infinito. Sea

$$
\phi(z)=\frac{e^{-z^2/2}}{\sqrt{2\pi}}
$$

la densidad normal estándar. Entonces

$$
\begin{aligned}
\mathbb E^{\mathbb Q^{T_2}}
\left[(F_{T_1}-K)^+\mid\mathcal F_t\right]
&=\int_{-d_2}^{\infty}
\left(F_t e^{-s^2/2+sz}-K\right)\phi(z)\,dz\\
&=F_t\int_{-d_2}^{\infty}
e^{-s^2/2+sz}\phi(z)\,dz
-K\int_{-d_2}^{\infty}\phi(z)\,dz.
\end{aligned}
$$

La integral del **strike** ya la calculamos. Como $K$ es fijo, su contribución es

$$
K\Phi(d_2).
$$

En la integral del **forward**, el valor de $F_{T_1}$ cambia con $z$. Para calcularla, juntamos el exponente del forward con el de la densidad y completamos el cuadrado:

$$
\begin{aligned}
e^{-s^2/2+sz}\phi(z)
&=\frac{1}{\sqrt{2\pi}}
\exp\left(-\frac{s^2}{2}+sz-\frac{z^2}{2}\right)\\
&=\frac{1}{\sqrt{2\pi}}
\exp\left(-\frac{(z-s)^2}{2}\right)\\
&=\phi(z-s).
\end{aligned}
$$

Con el cambio de variable $y=z-s$, el límite inferior pasa de $-d_2$ a $-d_2-s$. Por tanto,

$$
\begin{aligned}
\int_{-d_2}^{\infty}
e^{-s^2/2+sz}\phi(z)\,dz
&=\int_{-d_2-s}^{\infty}\phi(y)\,dy\\
&=\Phi(d_2+s).
\end{aligned}
$$

De aquí sale $d_1$. Al definir $d_1=d_2+s$, obtenemos

$$
d_1=\frac{\log(F_t/K)+s^2/2}{s}.
$$

**$\Phi(d_2)$ mide la probabilidad de que el caplet pague bajo la medida forward. El término $F_t\Phi(d_1)$ mide el valor esperado del fixing contando sólo los estados en que paga.**

Aparecen dos términos distintos porque en esos estados el strike siempre vale $K$, mientras que el fixing puede tomar distintos valores.

### 4. Obtener la prima en moneda

La esperanza que acabamos de calcular está expresada como diferencia de tasas. Multiplicamos por $N\delta$ para convertirla en un importe y por $P(t,T_2)$ para traerlo a la fecha de valuación:

$$
\boxed{
V_t=N\delta P(t,T_2)
\left[F_t\Phi(d_1)-K\Phi(d_2)\right].
}
$$

Cada factor responde a una parte del contrato. **$\tau=T_1-t$** determina cuánta incertidumbre queda hasta el fixing, **$\delta$** convierte la diferencia de tasas anualizadas en interés del periodo $[T_1,T_2]$ y **$P(t,T_2)$** descuenta hasta la fecha efectiva de pago.

# Floorlet



Sí. El **caplet** es una *call* sobre la tasa y el **floorlet** es una *put*. El floorlet paga si la tasa observada en $T_1$ cae por debajo del strike $K$. El comprador recibe en $T_2$:

$$
\boxed{
X_{T_2}^{\mathrm{floorlet}}
=N\delta(K-L_{T_1})^+.
}
$$

Si **recibes** intereses a tasa variable y compras un floorlet, el cobro combinado equivale a recibir la tasa

$$
L_{T_1}+(K-L_{T_1})^+
=\max(L_{T_1},K).
$$

Así, $K$ funciona como **piso** de la tasa que recibes, antes de considerar la prima.

Con el mismo numerario $P(t,T_2)$ y los mismos $d_1,d_2$ del caplet, la prima bajo **Black–76** es

$$
\boxed{
V_t^{\mathrm{floorlet}}
=N\delta P(t,T_2)
\left[K\Phi(-d_2)-F_t\Phi(-d_1)\right].
}
$$

Con el mismo strike y las mismas fechas, **caplet comprado menos floorlet comprado** da el flujo del FRA que paga fijo y recibe variable, con pago en $T_2$.

# Cap y floor: flujos y valuación completa con Black–76

Un **cap** protege contra subidas de la tasa de referencia durante varios periodos. Un **floor** protege contra bajadas. El cap está formado por una cartera de caplets y el floor por una cartera de floorlets. Todos los componentes tienen el **mismo strike $K$**, pero cada uno tiene su propio fixing y su propia fecha de pago.

### 1. Calendario y flujos

Sea $T_0<T_1<\cdots<T_n$ el calendario de devengo. Para el periodo $[T_{i-1},T_i]$, usaremos:

- $N$: nominal del contrato.
- $\delta_i$: fracción de año del periodo según la convención de días del contrato.
- $T_{f,i}$: fecha en que se observa la tasa de referencia del periodo, $L_i$.
- $T_{p,i}$: fecha de pago del cupón.
- $K$: tasa pactada como strike, común a todos los periodos.

En la convención simplificada que venimos usando, $T_{f,i}=T_{i-1}$ y $T_{p,i}=T_i$. Mantener los símbolos separados nos permite distinguir **cuándo se conoce el importe** de **cuándo se entrega el dinero**.

En cada fecha $T_{p,i}$, el comprador recibe

$$
\boxed{
X_i^{\mathrm{cap}}=N\delta_i(L_i-K)^+,
\qquad
X_i^{\mathrm{floor}}=N\delta_i(K-L_i)^+.
}
$$

Para quien **paga** intereses a tasa variable, el cap reduce el pago del periodo a

$$
N\delta_iL_i-X_i^{\mathrm{cap}}
=N\delta_i\min(L_i,K).
$$

Para quien **recibe** intereses a tasa variable, el floor eleva el cobro del periodo a

$$
N\delta_iL_i+X_i^{\mathrm{floor}}
=N\delta_i\max(L_i,K).
$$

En ambos casos, estamos hablando del interés neto **antes de considerar la prima**. El máximo se aplica **a cada periodo por separado**. Si la tasa queda por encima de $K$ en un periodo y por debajo en el siguiente, el cap cobra en el primero y su pago en el segundo es cero.

### 2. Valuación por ausencia de arbitraje

Consideremos dos portafolios:

- **A:** Compramos un cap.
- **B:** Compramos un caplet por cada periodo del cap, con el mismo nominal, strike, fixing y fecha de pago.

En cada $T_{p,i}$, ambos portafolios entregan exactamente $N\delta_i(L_i-K)^+$ en todos los estados. Por ausencia de arbitraje, tienen el mismo valor. El mismo argumento aplica al floor y su cartera de floorlets:

$$
\boxed{
V_t^{\mathrm{cap}}
=\sum_{i=1}^{n}V_{t,i}^{\mathrm{caplet}},
\qquad
V_t^{\mathrm{floor}}
=\sum_{i=1}^{n}V_{t,i}^{\mathrm{floorlet}}.
}
$$

Para cada componente, elegimos como numerario el bono que vence en **su propia fecha de pago** $T_{p,i}$. Denotamos por $P_i=P_d(t,T_{p,i})$ su precio hoy y por $\mathbb Q^{T_{p,i}}$ la medida forward asociada. Entonces, para un periodo pendiente de fijar,

$$
\begin{aligned}
V_{t,i}^{\mathrm{caplet}}
&=N\delta_iP_i\,
\mathbb E^{\mathbb Q^{T_{p,i}}}
\left[(L_i-K)^+\mid\mathcal F_t\right],\\
V_{t,i}^{\mathrm{floorlet}}
&=N\delta_iP_i\,
\mathbb E^{\mathbb Q^{T_{p,i}}}
\left[(K-L_i)^+\mid\mathcal F_t\right].
\end{aligned}
$$

**Cada opción se valúa con su propio bono de pago.** Una vez expresado cada precio en moneda de hoy, podemos sumarlos. Como el contrato es una suma de pagos individuales, para obtener su prima por este método no necesitamos especificar correlaciones entre los fixings.

### 3. Fórmulas completas de Black–76

Para el periodo $i$, sea $F_i(t)$ el **forward de la referencia de ese periodo**, $\sigma_i$ la volatilidad Black aplicable a ese componente y $\tau_i=T_{f,i}-t$ el tiempo en años hasta su fixing. Definimos

$$
s_i=\sigma_i\sqrt{\tau_i},
\qquad
d_{1,i}=
\frac{\ln(F_i(t)/K)+\tfrac12s_i^2}{s_i},
\qquad
d_{2,i}=d_{1,i}-s_i.
$$

Para $F_i(t)>0$, $K>0$ y $s_i>0$, al sustituir la fórmula de cada caplet obtenemos la **prima del cap**:

$$
\boxed{
V_t^{\mathrm{cap}}
=
N\sum_{i=1}^{n}
\delta_iP_d(t,T_{p,i})
\left[
F_i(t)\Phi(d_{1,i})
-K\Phi(d_{2,i})
\right].
}
$$

Análogamente, al sumar los floorlets obtenemos la **prima del floor**:

$$
\boxed{
V_t^{\mathrm{floor}}
=
N\sum_{i=1}^{n}
\delta_iP_d(t,T_{p,i})
\left[
K\Phi(-d_{2,i})
-F_i(t)\Phi(-d_{1,i})
\right].
}
$$

Aquí $\Phi$ es la función de distribución acumulada de la normal estándar. **$K$ es común a todo el contrato.** En cambio, el forward, el tiempo al fixing, la fracción de año y el factor de descuento se calculan para cada periodo. La volatilidad también puede variar por componente. Si te dan una **volatilidad plana del cap**, se usa el mismo valor de $\sigma_i$ en todos sus caplets para reproducir la prima del cap completo.

En el caso de **una sola curva**, con fixing al inicio y pago al final de cada periodo, el forward se obtiene como

$$
F_i(t)=\frac{1}{\delta_i}
\left[
\frac{P_d(t,T_{i-1})}{P_d(t,T_i)}-1
\right].
$$

Si usas curvas separadas, obtienes $F_i(t)$ de la **curva de proyección de la referencia** y mantienes $P_d(t,T_{p,i})$ de la **curva de descuento**.

Si un periodo **ya se fijó pero aún no se paga**, su aporte al precio es el flujo conocido multiplicado por $P_d(t,T_{p,i})$. Por ejemplo, para el cap es

$$
N\delta_iP_d(t,T_{p,i})(L_i-K)^+,
$$

usando el fixing observado. Los periodos ya liquidados se excluyen.

### 4. Paridad cap–floor y relación con el IRS

Compra un cap y vende un floor con el mismo nominal, strike, índice y calendario. En cada periodo, el flujo combinado es

$$
\begin{aligned}
X_i^{\mathrm{cap}}-X_i^{\mathrm{floor}}
&=N\delta_i\left[(L_i-K)^+-(K-L_i)^+\right]\\
&=N\delta_i(L_i-K).
\end{aligned}
$$

Es el cupón neto de un **IRS que paga fijo $K$ y recibe variable**, con las mismas fechas y fracciones en ambas patas. Por tanto, para un calendario de periodos aún sin fijar,

$$
\boxed{
V_t^{\mathrm{cap}}-V_t^{\mathrm{floor}}
=
N\sum_{i=1}^{n}
\delta_iP_d(t,T_{p,i})[F_i(t)-K]
=
V_t^{\mathrm{IRS\ pagador\ fijo}}.
}
$$

También se puede comprobar directamente en las fórmulas de Black–76 usando $\Phi(d)+\Phi(-d)=1$. Al restar caplet menos floorlet, los términos con $d_{1,i}$ suman $F_i(t)$ y los términos con $d_{2,i}$ suman $-K$. **La volatilidad desaparece de la diferencia** porque el flujo combinado vuelve a ser lineal.

En particular, si elegimos como strike la **tasa par del IRS con ese mismo calendario**,

$$
K_{\mathrm{par}}
=
\frac{
\displaystyle\sum_{i=1}^{n}
\delta_iP_d(t,T_{p,i})F_i(t)
}{
\displaystyle\sum_{i=1}^{n}
\delta_iP_d(t,T_{p,i})
},
$$

el IRS vale cero y **el cap y el floor tienen la misma prima**. Eso ocurre aunque los forwards individuales $F_i(t)$ sean distintos entre sí y ninguno coincida exactamente con $K_{\mathrm{par}}$.

```python
from math import erfc, log, sqrt


def precio_caplet_floorlet(
    nominal, delta, descuento, forward, strike, volatilidad, tau,
    tipo="caplet",
):
    """Calcula la prima Black–76 de un caplet o floorlet antes del fixing."""
    if tipo not in ("caplet", "floorlet"):
        raise ValueError("tipo debe ser 'caplet' o 'floorlet'.")

    if nominal <= 0 or delta <= 0 or descuento <= 0:
        raise ValueError("El nominal, delta y el descuento deben ser positivos.")

    if forward <= 0 or strike <= 0:
        raise ValueError("Black–76 estándar requiere forward y strike positivos.")

    if volatilidad < 0 or tau <= 0:
        raise ValueError("La volatilidad debe ser no negativa y tau debe ser positivo.")

    def Phi(x):
        """Distribución acumulada de la normal estándar."""
        return 0.5 * erfc(-x / sqrt(2))

    escala = nominal * delta * descuento
    s = volatilidad * sqrt(tau)

    if s == 0:
        diferencia = forward - strike if tipo == "caplet" else strike - forward
        return escala * max(diferencia, 0.0)

    d1 = (log(forward / strike) + 0.5 * s**2) / s
    d2 = d1 - s

    if tipo == "caplet":
        prima_en_tasa = forward * Phi(d1) - strike * Phi(d2)
    else:
        prima_en_tasa = strike * Phi(-d2) - forward * Phi(-d1)

    return escala * prima_en_tasa
```

```python
datos = dict(
    nominal=10_000_000,
    delta=90 / 360,    # Periodo que devenga intereses.
    descuento=0.96,    # Descuento hasta el pago.
    forward=0.084,
    strike=0.08,
    volatilidad=0.20,
    tau=0.50,         # Hoy faltan seis meses para el fixing.
)

caplet = precio_caplet_floorlet(**datos, tipo="caplet")
floorlet = precio_caplet_floorlet(**datos, tipo="floorlet")

print(f"Prima caplet:   {caplet:,.2f} MXN")
print(f"Prima floorlet: {floorlet:,.2f} MXN")
```

```python
def precio_cap_floor(nominal, strike, periodos, tipo="cap"):
    """Devuelve la prima total y el valor de cada componente."""
    if tipo not in ("cap", "floor"):
        raise ValueError("tipo debe ser 'cap' o 'floor'.")

    tipo_opcion = "caplet" if tipo == "cap" else "floorlet"

    primas = [
        precio_caplet_floorlet(
            nominal=nominal,
            strike=strike,
            tipo=tipo_opcion,
            **periodo,
        )
        for periodo in periodos
    ]

    return sum(primas), primas
```

```python
nominal = 10_000_000
strike = 0.08

periodos = [
    dict(delta=90 / 360, descuento=0.96, forward=0.084,
         volatilidad=0.20, tau=0.25),
    dict(delta=90 / 360, descuento=0.94, forward=0.087,
         volatilidad=0.21, tau=0.50),
    dict(delta=90 / 360, descuento=0.92, forward=0.090,
         volatilidad=0.22, tau=0.75),
]

prima_cap, caplets = precio_cap_floor(nominal, strike, periodos, tipo="cap")
prima_floor, floorlets = precio_cap_floor(nominal, strike, periodos, tipo="floor")

print(f"{'Periodo':>7} {'Caplet (MXN)':>16} {'Floorlet (MXN)':>16}")
for i, (caplet, floorlet) in enumerate(zip(caplets, floorlets), start=1):
    print(f"{i:>7} {caplet:>16,.2f} {floorlet:>16,.2f}")

print(f"\nPrima total del cap:   {prima_cap:,.2f} MXN")
print(f"Prima total del floor: {prima_floor:,.2f} MXN")
```

```python
def strike_atm_cap_floor(periodos):
    """Calcula el strike ATM común del cap y del floor."""
    suma_pesos = 0.0
    suma_ponderada = 0.0

    for periodo in periodos:
        peso = periodo["delta"] * periodo["descuento"]
        suma_pesos += peso
        suma_ponderada += peso * periodo["forward"]

    if suma_pesos <= 0:
        raise ValueError("La suma de los pesos debe ser positiva.")

    return suma_ponderada / suma_pesos
```

# Swaption


Una **swaption europea** da a su comprador el **derecho, pero no la obligación**, de entrar en un IRS en una fecha de ejercicio pactada $T_e$, a una tasa fija $K$.

- Una **payer swaption** da derecho a **pagar fijo a $K$ y recibir variable**.
- Una **receiver swaption** da derecho a **recibir fijo a $K$ y pagar variable**.

Consideraremos una swaption con **entrega física**. Si el comprador ejerce en $T_e$, entra en el IRS, que comienza en $T_0\geq T_e$. A partir de ahí, los cupones del IRS se pagan según su calendario. El *payoff* de la swaption en $T_e$ es el **valor de mercado del IRS que puedes recibir al ejercer**, o cero si decides dejar vencer la opción.

### 1. La anualidad y la tasa par del IRS

Sea $N$ el nominal del IRS, $T_1,\ldots,T_n$ las fechas de pago de la pata fija y $\delta_i$ la fracción de año de cada cupón. Escribimos $P(u,T_i)$ para el precio en $u$ de un bono cupón cero que paga una unidad monetaria en $T_i$.

La pata fija paga $NK\delta_i$ en cada $T_i$. Por tanto, su valor presente en $u$ es

$$
V_u^{\mathrm{fija}}
=
\sum_{i=1}^{n}NK\delta_iP(u,T_i)
=
NK\underbrace{\sum_{i=1}^{n}\delta_iP(u,T_i)}_{A_u}.
$$

La **anualidad** es $A_u=\sum_i\delta_iP(u,T_i)$. La fracción $\delta_i$ convierte la tasa anual en interés del periodo y $P(u,T_i)$ trae ese pago a $u$. La suma abarca **todos los pagos de la pata fija**, sean trimestrales, semestrales o anuales. También puedes interpretar $A_u$ como el precio de una cartera que contiene $\delta_i$ unidades del bono que vence en cada $T_i$. Si multiplicas esa cartera por $NK$, reproduce todos los cupones fijos.

Sea $B_u$ el valor presente de la **pata variable por unidad de nominal**. En una sola curva, cuando cada cupón variable se paga al final de su periodo,

$$
B_u=P(u,T_0)-P(u,T_n).
$$

El IRS que paga fijo a $K$ vale $N(B_u-KA_u)$. La **tasa par del IRS** es la tasa fija $S_u$ que hace que ese valor sea cero:

$$
B_u-S_uA_u=0
\qquad\Longrightarrow\qquad
\boxed{S_u=\frac{B_u}{A_u}.}
$$

La llamamos **tasa swap forward** porque la observamos en $u$ para un IRS que empieza en $T_0$, después de $u$ o en esa misma fecha. Si las patas comparten fechas y fracciones, también se puede escribir como

$$
S_u=
\frac{\displaystyle\sum_i\delta_iP(u,T_i)F_i(u)}
{\displaystyle\sum_i\delta_iP(u,T_i)}.
$$

Aquí $F_i(u)$ es el forward del **periodo $i$** y $S_u$ es la tasa par de **todo el IRS**. Los $F_i(u)$ ayudan a construir $S_u$. Para valuar la swaption con Black–76, modelamos directamente cómo se mueve **$S_u$** hasta el ejercicio.

### 2. Payoff en la fecha de ejercicio

Sustituyendo $B_u=A_uS_u$, el valor del IRS que **paga fijo** al strike $K$ es

$$
V_u^{\mathrm{IRS,payer}}(K)=NA_u(S_u-K).
$$

En $T_e$, si $S_{T_e}>K$, la payer te permite pagar una tasa fija inferior a la que el mercado exigiría para entrar al mismo IRS a la par. Ejercer te da un IRS con valor positivo. Si $S_{T_e}<K$, puedes dejar vencer la opción y evitar recibir un IRS con valor negativo. Para la receiver, la desigualdad se invierte.

Por tanto, el **valor de ejercicio** es

$$
\boxed{
\begin{aligned}
X_{T_e}^{\mathrm{payer}}
&=NA_{T_e}(S_{T_e}-K)^+,\\
X_{T_e}^{\mathrm{receiver}}
&=NA_{T_e}(K-S_{T_e})^+.
\end{aligned}
}
$$

El IRS tiene un valor que puede ser positivo o negativo. La swaption toma **la parte positiva de ese valor completo**. Ésa es también la diferencia con un cap: en el cap aplicas el máximo a cada cupón; en la swaption lo aplicas una sola vez al valor del IRS.

### 3. Por qué usamos la anualidad como numerario

Usar $A_u$ como numerario significa **medir los precios en copias de la cartera de bonos que reproduce la pata fija por unidad de tasa y nominal**. Si esa cartera cuesta $A_u$, un contrato que vale $V_u$ unidades monetarias equivale a $V_u/A_u$ copias de la cartera.

La cartera tiene valor positivo y sus bonos pagan después de $T_e$, por lo que podemos usarla como numerario hasta el ejercicio. Sea $M_u$ la cuenta monetaria y $\mathbb Q$ su medida neutral al riesgo. La medida asociada a la anualidad, $\mathbb Q^A$, se define mediante

$$
Z_u=
\left.\frac{d\mathbb Q^A}{d\mathbb Q}\right|_{\mathcal F_u}
=
\frac{A_u/M_u}{A_0/M_0}.
$$

Como $A_u/M_u$ es una martingala positiva, $Z_u$ repondera los estados al pasar de la cuenta monetaria a la anualidad. Para ver qué le pasa a la tasa swap, recuerda que en una sola curva $B_u$ es el precio de la cartera que compra el bono de $T_0$ y vende el de $T_n$. Hasta $T_e$, valorando el bono de $T_0$ justo antes de su pago si $T_e=T_0$, $B_u/M_u$ es una martingala bajo $\mathbb Q$. Por Bayes condicional,

$$
\begin{aligned}
\mathbb E^{\mathbb Q^A}\!\left[
\frac{B_u}{A_u}\,\middle|\,\mathcal F_t
\right]
&=
\frac{1}{Z_t}
\mathbb E^{\mathbb Q}\!\left[
Z_u\frac{B_u}{A_u}\,\middle|\,\mathcal F_t
\right]\\
&=
\frac{1}{Z_t(A_0/M_0)}
\mathbb E^{\mathbb Q}\!\left[
\frac{B_u}{M_u}\,\middle|\,\mathcal F_t
\right]\\
&=\frac{B_t}{A_t}=S_t.
\end{aligned}
$$

Así, **$S_u=B_u/A_u$ es una martingala bajo $\mathbb Q^A$**. La misma elección de numerario simplifica el payoff: en cada escenario, dividir $X_{T_e}^{\mathrm{payer}}$ entre $A_{T_e}$ deja $N(S_{T_e}-K)^+$. Después tomamos la esperanza bajo $\mathbb Q^A$ y multiplicamos por $A_t$ para volver a moneda de hoy:

$$
\boxed{
\begin{aligned}
V_t^{\mathrm{payer}}
&=
NA_t\,\mathbb E^{\mathbb Q^A}
\left[(S_{T_e}-K)^+\mid\mathcal F_t\right],\\
V_t^{\mathrm{receiver}}
&=
NA_t\,\mathbb E^{\mathbb Q^A}
\left[(K-S_{T_e})^+\mid\mathcal F_t\right].
\end{aligned}
}
$$

El punto es que **$A_{T_e}$ es aleatoria y se cancela dentro del payoff expresado en unidades del numerario**. El factor que multiplica la esperanza es $A_t$, el precio de esa cartera **hoy**.

### 4. Resolver las esperanzas con Black–76

Black–76 supone que la tasa swap forward tiene volatilidad relativa constante $\sigma$ bajo la medida de la anualidad:

$$
dS_u=\sigma S_u\,dW_u^A.
$$

Por Itô,

$$
d\log S_u=-\frac12\sigma^2\,du+\sigma\,dW_u^A.
$$

Si $\tau=T_e-t$ es el tiempo en años hasta el ejercicio y $s=\sigma\sqrt{\tau}$, al integrar obtenemos

$$
S_{T_e}=S_t e^{-s^2/2+sZ},
\qquad
Z=\frac{W_{T_e}^A-W_t^A}{\sqrt{\tau}}
\sim\mathcal N(0,1)
$$

bajo $\mathbb Q^A$, condicionalmente a $\mathcal F_t$. Para $S_t,K,s>0$, definimos

$$
d_2=\frac{\ln(S_t/K)-s^2/2}{s},
\qquad
d_1=d_2+s.
$$

**Payer.** Despejar $Z$ en $S_t e^{-s^2/2+sZ}>K$ da $Z>-d_2$. Si $\phi$ es la densidad normal estándar,

$$
\begin{aligned}
\mathbb E^{\mathbb Q^A}[
(S_{T_e}-K)^+\mid\mathcal F_t]
&=
S_t\int_{-d_2}^{\infty}
e^{-s^2/2+sz}\phi(z)\,dz
-K\int_{-d_2}^{\infty}\phi(z)\,dz\\
&=S_t\Phi(d_1)-K\Phi(d_2).
\end{aligned}
$$

La integral del strike es $\Phi(d_2)$ por simetría. Para la integral de la tasa, completar el cuadrado da $e^{-s^2/2+sz}\phi(z)=\phi(z-s)$. Con el cambio $y=z-s$, el límite inferior pasa a $-d_2-s=-d_1$, de modo que esa integral vale $\Phi(d_1)$.

**Receiver.** Ahora se ejerce si $S_{T_e}<K$, es decir, si $Z<-d_2$. Usamos la misma identidad, pero integramos hacia abajo:

$$
\begin{aligned}
\mathbb E^{\mathbb Q^A}[
(K-S_{T_e})^+\mid\mathcal F_t]
&=
K\int_{-\infty}^{-d_2}\phi(z)\,dz
-S_t\int_{-\infty}^{-d_2}\phi(z-s)\,dz\\
&=K\Phi(-d_2)-S_t\Phi(-d_1).
\end{aligned}
$$

Al multiplicar por $NA_t$, obtenemos las **primas en $t$**:

$$
\boxed{
\begin{aligned}
V_t^{\mathrm{payer}}
&=
NA_t\left[S_t\Phi(d_1)-K\Phi(d_2)\right],\\
V_t^{\mathrm{receiver}}
&=
NA_t\left[K\Phi(-d_2)-S_t\Phi(-d_1)\right].
\end{aligned}
}
$$

Como comprobación, **comprar una payer y vender una receiver** con las mismas condiciones reproduce el valor del IRS pagador fijo. En $T_e$, la diferencia de payoffs es $NA_{T_e}(S_{T_e}-K)$. Por ausencia de arbitraje,

$$
V_t^{\mathrm{payer}}-V_t^{\mathrm{receiver}}
=
NA_t(S_t-K).
$$

Por eso el **strike ATM es $K=S_t$**, la tasa par de hoy del IRS subyacente. A ese strike, el IRS vale cero y ambas swaptions tienen la misma prima.

```python
def datos_swap_forward(fracciones_fijas, descuentos_pago, descuento_inicio):
    """Devuelve la anualidad y la tasa par del IRS en una sola curva."""
    if len(fracciones_fijas) == 0 or len(fracciones_fijas) != len(descuentos_pago):
        raise ValueError("Debe haber una fracción fija por cada fecha de pago.")
    if any(delta <= 0 for delta in fracciones_fijas):
        raise ValueError("Las fracciones de año deben ser positivas.")
    if descuento_inicio <= 0 or any(p <= 0 for p in descuentos_pago):
        raise ValueError("Los factores de descuento deben ser positivos.")

    anualidad = sum(
        delta * descuento
        for delta, descuento in zip(fracciones_fijas, descuentos_pago)
    )
    valor_variable = descuento_inicio - descuentos_pago[-1]
    tasa_swap = valor_variable / anualidad

    return anualidad, tasa_swap
```

```python
from math import erfc, log, sqrt


def precio_swaption(nominal, anualidad, tasa_swap, strike, volatilidad, tau,
                    tipo="payer"):
    """Calcula la prima Black–76 de una swaption payer o receiver."""
    if tipo not in ("payer", "receiver"):
        raise ValueError("tipo debe ser 'payer' o 'receiver'.")
    if nominal <= 0 or anualidad <= 0:
        raise ValueError("El nominal y la anualidad deben ser positivos.")
    if volatilidad < 0 or tau < 0:
        raise ValueError("La volatilidad y tau deben ser no negativos.")

    escala = nominal * anualidad
    s = volatilidad * sqrt(tau)

    if s == 0:
        diferencia = tasa_swap - strike if tipo == "payer" else strike - tasa_swap
        return escala * max(diferencia, 0.0)

    if tasa_swap <= 0 or strike <= 0:
        raise ValueError("Black–76 estándar requiere tasa swap y strike positivos.")

    def Phi(x):
        return 0.5 * erfc(-x / sqrt(2))

    d1 = (log(tasa_swap / strike) + 0.5 * s**2) / s
    d2 = d1 - s

    if tipo == "payer":
        prima_en_tasa = tasa_swap * Phi(d1) - strike * Phi(d2)
    else:
        prima_en_tasa = strike * Phi(-d2) - tasa_swap * Phi(-d1)

    return escala * prima_en_tasa
```
